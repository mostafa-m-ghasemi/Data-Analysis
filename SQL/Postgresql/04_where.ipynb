# WHERE & Filtering Data

```python
import psycopg
import pandas as pd
conn = psycopg.connect(
    host = 'localhost',
    port= 5432,
    dbname ='chinook',
    user = 'postgres',
    password = '231165'
)

cur = conn.cursor()
```

## Basic Filtering with WHERE

```python
# Filter customers by country

result = pd.read_sql('''
SELECT customer_id, first_name, last_name, country
FROM customer
WHERE country = 'USA'
''', conn)

result
```

```python
# Filter tracks with a price above 1$

result = pd.read_sql('''
SELECT track_id, name, unit_price
FROM track
WHERE unit_price > 1
''', conn)

result
```

## Filtering wuth AND

```python
# Apply multiole conditions with AND

result = pd.read_sql('''
SELECT  name, unit_price, milliseconds
FROM track
WHERE unit_price > 1 AND milliseconds > 300000
''', conn)

result
```

## Filtering with OR

```python
# Find customers from USAA or Canada

result = pd.read_sql('''
SELECT customer_id, first_name, last_name, country
FROM customer
WHERE country = 'USA' OR country = 'Canada'
''', conn)

result
```

## Filtering with BETWEEN

```python
result = pd.read_sql('''
SELECT invoice_id, customer_id, total
FROM invoice
WHERE total BETWEEN 5 AND 10
''', conn)

result
```

## Filtering with IN

```python
# Filter using a list of values

result = pd.read_sql('''
SELECT first_name, last_name, country
FROM customer
WHERE country IN ('USA', 'Canada', 'Brazil')
''', conn)

result
```

## Filtering with LIKE

```python
# Find names starting with a letter

result = pd.read_sql('''
SELECT first_name, last_name
FROM customer
WHERE first_name LIKE 'A%'
''', conn)

result
```

```python
# Find names ending with a letter

result = pd.read_sql('''
SELECT first_name, last_name
FROM customer
WHERE first_name LIKE '%a'
''', conn)

result
```

```python
# Perform case-insensitive matching ---> ILIKE

result = pd.read_sql('''
SELECT first_name, last_name
FROM customer
WHERE first_name ILIKE 'a%'
''', conn)

result
```

## Filtering NULL Values

```python
# The ccustomers without a company

result = pd.read_sql('''
SELECT first_name, last_name, company
FROM customer
WHERE company IS NULL
''', conn)

result
```

```python
# The customers with a company

result = pd.read_sql('''
SELECT first_name, last_name, company
FROM customer
WHERE company IS NOT NULL
''', conn)

result
```

## Combine Filtering with COUNT

```python
result = pd.read_sql('''
SELECT COUNT(*)
FROM customer
WHERE country = 'USA'
''', conn)

result
```
