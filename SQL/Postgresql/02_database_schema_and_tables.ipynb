## Explore the Database

## List all Tables

```python
cur.execute('''
SELECT table_name
FROM information_schema.tables
WHERE table_schema = 'public'
ORDER BY table_name;
''')
tables = cur.fetchall()

tables
```

## Get Columns Information

```python
cur.execute('''
SELECT column_name
FROM information_schema.columns
WHERE table_schema = 'public'
AND table_name = 'artist'
ORDER BY table_name;
''')
columns = cur.fetchall()

columns
```
