# SELECT & Selecting Columns

```python
import psycopg
import pandas as pd
conn = psycopg.connect(
    host = 'localhost',
    port= 5432,
    dbname ='chinook',
    user = 'postgres',
    password = '231165'
)
```

```python
cur = conn.cursor()
```

## Select All Culumns

```python
result = pd.read_sql('''
SELECT *
FROM artist
''', conn)

result
```

## SELECT Specific Columns

```python

result = pd.read_sql('''
SELECT name
FROM artist
''', conn)

result
```

## Rename Columns with AS

```python
result = pd.read_sql('''
SELECT artist_id AS ID, name AS Name
FROM artist
''', conn)

result
```

## SELECT Distinct Values

```python
# Select unique values

result = pd.read_sql('''
SELECT DISTINCT country
FROM customer
''', conn)

result
```

## Combine Multiple Columns

```python
result = pd.read_sql('''
SELECT first_name, last_name, email
FROM customer
''', conn)

result
```

## Count Rows

```python
result = pd.read_sql('''
SELECT COUNT(*)
FROM customer
''', conn)

result
```
