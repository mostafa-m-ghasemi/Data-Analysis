# Aggregate Functions

```python
import psycopg
import pandas as pd
conn = psycopg.connect(
    host = 'localhost',
    port= 5432,
    dbname ='chinook',
    user = 'postgres',
    password = '231165'
)

cur = conn.cursor()
```

## COUNT()

```python
# Count all tracks

result = pd.read_sql('''
SELECT COUNT(*) AS total_tracks
FROM track
''', conn)

result
```

## COUNT(*) vs COUNT(column)

```python
# Compare total rows with not-NULL values

result = pd.read_sql('''
SELECT COUNT(*) AS total_rows, COUNT(composer) AS track_with_composer
FROM track
''', conn)

result
```

## SUM()

```python
result = pd.read_sql('''
SELECT SUM(total) AS total_revenue
FROM invoice
''', conn)

result
```

## AVG()

```python
result = pd.read_sql('''
SELECT AVG(total) AS total_average
FROM invoice
''', conn)

result
```

## MIN() and MAX()

```python
result = pd.read_sql('''
SELECT MIN(unit_price) AS min_price, MAX(unit_price) AS max_price
FROM track
''', conn)

result
```

```python
result = pd.read_sql('''
SELECT MIN(invoice_date) AS earliest, MAX(invoice_date) AS last
FROM invoice
''', conn)

result
```

## Combine Aggregate Finctions

```python
result = pd.read_sql('''
SELECT 
COUNT(*) AS invoice_count,
SUM(total) AS total_revenue,
AVG(total) AS total_avg,
MIN(total) AS min_invoice
FROM invoice
''', conn)

result
```
