# ORDER BY & LIMIT

```python
import psycopg
import pandas as pd
conn = psycopg.connect(
    host = 'localhost',
    port= 5432,
    dbname ='chinook',
    user = 'postgres',
    password = '231165'
)

cur = conn.cursor()
```

## Introduction to ORDER BY

### Basic ORDER BY

```python
result = pd.read_sql('''
SELECT name
FROM track
ORDER BY name
''', conn)

result
```

```python
result = pd.read_sql('''
SELECT name
FROM track
ORDER BY name DESC
''', conn)

result
```

### Sorting numbers

```python
result = pd.read_sql('''
SELECT name, unit_price
FROM track
ORDER BY unit_price ASC
''', conn)

result
```

### Sorting Dates

```python
result = pd.read_sql('''
SELECT invoice_date, total
FROM invoice
ORDER BY invoice_date DESC
''', conn)

result
```

## Sorting by Multiple Columns

```python
result = pd.read_sql('''
SELECT name, unit_price
FROM track
ORDER BY unit_price DESC, name ASC
''', conn)

result
```

## Introduction to LIMIT

```python
result = pd.read_sql('''
SELECT name, unit_price
FROM track
ORDER BY unit_price DESC, name ASC
LIMIT 10
''', conn)

result
```

## Using OFFSET for Pagination

```python
result = pd.read_sql('''
SELECT name, unit_price
FROM track
ORDER BY unit_price DESC, name ASC
LIMIT 10 OFFSET 5
''', conn)

result
```

## Handling NULL Values in ORDER BY

```python
result = pd.read_sql('''
SELECT name
FROM track
ORDER BY name ASC NULLS FIRST
''', conn)

result
```

```python
result = pd.read_sql('''
SELECT name
FROM track
ORDER BY name ASC NULLS LAST
''', conn)

result
```
