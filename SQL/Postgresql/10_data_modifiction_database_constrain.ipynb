# Data Modifiction & Database Constrains

```python
import psycopg
import pandas as pd
conn = psycopg.connect(
    host = 'localhost',
    port= 5432,
    dbname ='chinook',
    user = 'postgres',
    password = '231165'
)

cur = conn.cursor()
```

## INSERT - Adding Data

```python
# Inser new record to a table

query = '''
INSERT INTO genre (genre_id, name)
VALUES (1000, 'Test Genre')
'''

with conn.cursor() as cur:
    cur.execute(query)

conn.commit()
```

## UPDATE - Modifyung Data

```python
# Update the test genre

query = '''
UPDATE genre 
SET name = 'Updated Test Genre'
WHERE genre_id = 1000
'''

with conn.cursor() as cur:
    cur.execute(query)

conn.commit()
```

```python
query = '''
SELECT *
FROM genre
WHERE genre_id = 1000
'''

df = pd.read_sql_query(query, conn)

df
```

## DELETE - Removing Data

```python
query = '''
DELETE FROM genre
WHERE genre_id = 1000
'''

with conn.cursor() as cur:
    cur.execute(query)

conn.commit()
```
