# WHERE

```python
import sqlite3
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Filter with a Condition

```python
# SELECT columns
# FROM table
# WHERE condition;
```

```python
cursor.execute("""
SELECT *
FROM tracks
WHERE GenreId = 1
""")
result = cursor.fetchall()
result[:20]
```

## Numeric Comparisons

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
WHERE milliseconds > 300000
""")
result = cursor.fetchall()
result[:20]
```

## Multiple Conditions with AND

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
WHERE GenreId = 1
AND milliseconds > 300000
""")
result = cursor.fetchall()
result[:20]
```

## Multiple Conditions with OR

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
WHERE GenreId = 1
OR GenreId = 2
""")
result = cursor.fetchall()
result[:20]
```

## NOT

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
WHERE GenreId != 1
""")
result = cursor.fetchall()
result[:20]
```

## Text Filtering

```python
cursor.execute("""
SELECT *
FROM artists
WHERE name = 'CD/DC'
""")
result = cursor.fetchall()
result[:20]
```

## LIKE

```python
cursor.execute("""
SELECT *
FROM artists
WHERE name LIKE 'A%'
""")
result = cursor.fetchall()
result[:20]
```

```python
# 'A%'     --> starts with A
# '%A'     --> ends with A
# '%rock%' --> contains rock
```

```python
cursor.execute("""
SELECT *
FROM artists
WHERE name LIKE '%rock%'
""")
result = cursor.fetchall()
result[:20]
```

## Combine WHERE, ORDER BY, and LIMIT

```python
# Filter, sort, and limit results

cursor.execute("""
SELECT name, milliseconds
FROM tracks
WHERE GenreId = 1
ORDER BY milliseconds DESC
LIMIT 10
""")
result = cursor.fetchall()
result[:20]
```

## WHERE with Pandas

```python
import pandas as pd
df = pd.read_sql_query("""
SELECT name, milliseconds
FROM tracks
WHERE GenreId = 1
ORDER BY milliseconds DESC
LIMIT 10""", conn)

df
```
