# CREATE Table & Database Structure

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Inspect Existing Tables

```python
pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
""", conn)
```

## CREATE TABLE

```python
cursor.execute('''
CREATE TABLE practice_data(
id INTEGER,
name TEXT,
age INTEGER
)
''')
```

## Check the New Table

```python
pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
""", conn)
```

## SQLite Data Types

```python
# INTEGER --> int
# REAL    --> float
# TEXT    --> str
# BLOB    --> 0 & 1
# NULL    --> empty
```

## PRIMARY KEY

```python
cursor.execute('''
CREATE TABLE students(
studant_id INTEGER PRIMARY KEY,
name TEXT,
score REAL
)
''')
```

```python
pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
""", conn)
```

## AUTOINCREMENT

```python
# No need to add id!
```

```python
cursor.execute('''
CREATE TABLE courses (
course_id INTEGER PRIMARY KEY AUTOINCREMENT,
course_name TEXT,
duration INTEGER
)
''')
```

```python
cursor.execute('''
INSERT INTO courses (course_name, duration)
VALUES (?, ?)
''', ('Pathon',40 ))
```

```python
pd.read_sql('''
SELECT *
FROM courses
''', conn)
```

## NOT NULL

```python
cursor.execute('''
CREATE TABLE products(
product_id INTEGER PRIMARY KEY,
product_name TEXT NOT NULL,
price REAL NOT NULL
)
''')
```

## UNIQUE

```python
cursor.execute('''
CREATE TABLE users(
user_id INTEGER PRIMARY KEY,
username TEXT UNIQE
)
''')
```

```python
cursor.execute('''
INSERT INTO users (username)
VALUES (?)
''', ('mostafa',))

conn.commit()
```

```python
cursor.execute('''
INSERT INTO users (username)
VALUES (?)
''', ('mostafa',))
conn.commit()
```

## DEFAULT

```python
cursor.execute('''
CREATE TABLE tasks (
task_id INTEGER PRIMARY KEY,
yask_name TEXT NOT NULL,
status TEXT DEFAULT 'Pending'
)
''')

conn.commit()
```

```python
cursor.execute('''
INSERT INTO tasks (yask_name)
VALUES(?)
''', ('LEARN',))
conn.commit()
```

```python
pd.read_sql('''
SELECT *
FROM tasks
''', conn)
```

## Foreign Key

```python
cursor.execute('''
CREATE TABLE departments (
dep_id INTEGER PRIMARY KEY,
dep_name TEXT
)
''')

cursor.execute('''
CREATE TABLE employee (
em_id INTEGER PRIMARY KEY,
em(name TEXT NUT NOLL,
dep_id INTEGER,
FOREIGN KEY (dep_id)
REFRENCES (dep_id)
)
''')

conn.commit()
```
