# SELECT

```python
import sqlite3
```

```python
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Select all columns

```python
cursor.execute("""
SELECT *
FROM artists
""")
result = cursor.fetchall()
result
```

## Select specific columns

```python
cursor.execute("""
SELECT name
FROM artists
""")
result = cursor.fetchall()
result[:10]
```

## LIMIT

```python
# Limit the number of rows
cursor.execute("""
SELECT name
FROM artists
LIMIT 5
""")
result = cursor.fetchall()
result
```

## DISTINCT

```python
# Select unique values
cursor.execute("""
SELECT DISTINCT GenreId
FROM tracks
""")
result = cursor.fetchall()
result
```

## Select with Pandas

```python
# Load SQL results into a DataFrame
import pandas as pd

df = pd.read_sql_query("""
SELECT ArtistId, name
FROM artists
LIMIT 20""", conn)

df
```

```python
#  Select all columns with Pandas
df = pd.read_sql_query("""
SELECT *
FROM artists
LIMIT 20""", conn)

df
```
