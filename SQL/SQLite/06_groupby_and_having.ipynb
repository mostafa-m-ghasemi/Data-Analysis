# GROUP BY & HAVING

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Basic GROUP BY

```python
# Group rows and count each group
cursor.execute('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
GROUP BY GenreId
''')
result = cursor.fetchall()
result
```

## GROUP BY with SUM

```python
# Calculate the total foreach group
cursor.execute('''
SELECT GenreId, SUM(milliseconds) AS total_time
FROM tracks
GROUP BY GenreId
''')
result = cursor.fetchall()
result
```

## GROUP BY with AVG

```python
# Calculate the average for each group
cursor.execute('''
SELECT GenreId, AVG(milliseconds) AS average_time
FROM tracks
GROUP BY GenreId
''')
result = cursor.fetchall()
result
```

## GROUP BY wuth Multiple Aggregations

```python
# Calculate multiple statistics per group
cursor.execute('''
SELECT GenreId,
AVG(milliseconds) AS average_time,
MIN(milliseconds) AS shortest_track,
MAX(milliseconds) AS longest_trsck
FROM tracks
GROUP BY GenreId
''')
result = cursor.fetchall()
result
```

## GROUP BY wuth ORDER BY

```python
# Sort groups by their count
cursor.execute('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
GROUP BY GenreId
ORDER BY track_count DESC
''')
result = cursor.fetchall()
result
```

## HAVING

```python
# Filter grouped results
cursor.execute('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
GROUP BY GenreId
HAVING COUNT(*) > 100
ORDER BY track_count DESC
''')
result = cursor.fetchall()
result
```

## HAVING vs WHERE

```python
# WHERE    --> Filter rows
# HAVING   --> Filter groups
```

## GROUP BY wuth WHERE

```python
# Filter rows before grouping
cursor.execute('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
WHERE milliseconds > 300000
GROUP BY GenreId
ORDER BY track_count DESC
''')
result = cursor.fetchall()
result
```

## WHERE + GROUP BY + HAVING + ORDER BY

```python
# Filter, group, aggregate and sortt
cursor.execute('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
WHERE milliseconds > 300000
GROUP BY GenreId
HAVING COUNT(*) > 50
ORDER BY track_count DESC
''')
result = cursor.fetchall()
result
```

## GROUP BY with Pandas

```python
df = pd.read_sql_query('''
SELECT GenreId, COUNT(*) AS track_count
FROM tracks
WHERE milliseconds > 300000
GROUP BY GenreId
HAVING COUNT(*) > 50
ORDER BY track_count DESC
''', conn)

df
```
