# Subqueries

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Basic Subquery

```python
# Find values above the overall average
cursor.execute('''
SELECT Name, milliseconds
FROM tracks
WHERE milliseconds > (
SELECT AVG(milliseconds)
FROM tracks
)
''')

result = cursor.fetchall()
result
```

## Subquery with MAX()

```python
cursor.execute('''
SELECT Name, milliseconds
FROM tracks
WHERE milliseconds = (
SELECT MAX(milliseconds)
FROM tracks
)
''')

result = cursor.fetchall()
result
```

## Subquery with MIN()

```python
cursor.execute('''
SELECT Name, milliseconds
FROM tracks
WHERE milliseconds = (
SELECT MIN(milliseconds)
FROM tracks
)
''')

result = cursor.fetchall()
result
```

## Subquery with IN()

```python
cursor.execute('''
SELECT title, ArtistId
FROM albums
WHERE ArtistId IN (
SELECT ArtistId
FROM artists
WHERE name LIKE 'A%'
)
''')

result = cursor.fetchall()
result
```

## Subquery with COUNT()

```python
cursor.execute('''
SELECT name
FROM artists
WHERE ArtistId IN (
SELECT ArtistId
FROM albums
GROUP BY ArtistId
HAVING COUNT(*) > 5
)
''')

result = cursor.fetchall()
result
```
