```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## INSERT - Add New Data

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 5
''', conn)
```

## INSERT a New Artist

```python
cursor.execute('''
INSERT INTO artists (name)
VALUES (?)
''', ('Test Artist', ))

conn.commit()
```

## Verify INSERT

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 5
''', conn)
```

## INSERT Multipe Rows

```python
artist_name = [('Test Artist 2',),('Test Artist 3',),('Test Artist 4',)]

cursor.executemany('''
INSERT INTO artists (name)
VALUES (?)
''', artist_name)

conn.commit()
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```

## Update - Modify Existing Data

```python
cursor.execute("""
UPDATE artists
SET name = ?
WHERE name = ?
""", ("Updated Artist", "Test Artist"))

conn.commit()
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```

## DELETE - Remove Data

```python
cursor.execute("""
DELETE FROM artists
WHERE name = ?
""", ("Updated Artist",))

conn.commit()
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```

## DELETE Multiple Rows

```python
cursor.execute("""
DELETE FROM artists
WHERE name LIKE 'Test Artist%'
""")

conn.commit()
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```

## Rollback

```python
cursor.executemany('''
INSERT INTO artists (name)
VALUES (?)
''', artist_name)
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```

```python
cursor.executemany('''
INSERT INTO artists (name)
VALUES (?)
''', artist_name)

conn.rollback()
```

```python
pd.read_sql('''
SELECT *
FROM artists
ORDER BY ArtistId DESC
LIMIT 10
''', conn)
```
