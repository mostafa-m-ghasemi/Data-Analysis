# Aggregate Functions

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## COUNT()

```python
cursor.execute('''
SELECT COUNT(*)
FROM tracks;
''')
result = cursor.fetchone()
result
```

## COUNT() with a Column

```python
cursor.execute('''
SELECT COUNT(Name)
FROM tracks;
''')
result = cursor.fetchone()
result
```

```python
# COUNT(*)       --> count all rows
# COUNT(Column)  --> count all values not null
```

## SUM()

```python
# Calculate the total

cursor.execute('''
SELECT SUM(milliseconds)
FROM tracks;
''')
result = cursor.fetchone()
result
```

## AVG()

```python
# Calculate the average

cursor.execute('''
SELECT AVG(milliseconds)
FROM tracks;
''')
result = cursor.fetchone()
result
```

## MIN()

```python
# Find the minimum value

cursor.execute('''
SELECT MIN(milliseconds)
FROM tracks;
''')
result = cursor.fetchone()
result

```

## MAX()

```python
# Find the maximum value

cursor.execute('''
SELECT MAX(milliseconds)
FROM tracks;
''')
result = cursor.fetchone()
result

```

## Using Multiple Functions

```python
cursor.execute('''
SELECT 
COUNT(*) AS track_count,
SUM(milliseconds) AS total_time,
AVG(milliseconds) AS average_time,
MAX(milliseconds) AS shortest_track,
MAX(milliseconds) AS longest_track
FROM tracks;
''')
result = cursor.fetchone()
result
```

## Aggregate Functions with Pandas

```python
df = pd.read_sql_query('''
SELECT 
COUNT(*) AS track_count,
SUM(milliseconds) AS total_time,
AVG(milliseconds) AS average_time,
MAX(milliseconds) AS shortest_track,
MAX(milliseconds) AS longest_track
FROM tracks;
''', conn)

df
```
