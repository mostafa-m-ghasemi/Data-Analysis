# JOIN

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Understanding the Relationship

```python
# artist table --> arist_id, name
# album table  --> album_id, title, artisr_id
```

## Basic INNER JOIN

```python
# Combine two related rows from two tables
cursor.execute('''
SELECT albums.Title, artists.Name
FROM albums
INNER JOIN artists
ON  albums.ArtistId = artists.ArtistId

''')

result = cursor.fetchall()
result
```

## Using Table Aliases

```python
# Use aliases for shorter table name

cursor.execute('''
SELECT a.Title, ar.Name
FROM albums AS a
INNER JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId

''')
# albums   --> a
# artistrs --> ar

result = cursor.fetchall()
result
```

## JOIN with WHERE

```python
# Filter joined results

cursor.execute('''
SELECT a.Title, ar.Name
FROM albums AS a
INNER JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId
WHERE ar.name = 'AC/DC'
''')
# albums   --> a
# artistrs --> ar

result = cursor.fetchall()
result
```

## JOIN with ORDER BY

```python
# Sort joined results

cursor.execute('''
SELECT a.Title, ar.Name
FROM albums AS a
INNER JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId
ORDER BY ar.name
''')
# albums   --> a
# artistrs --> ar

result = cursor.fetchall()
result
```

## JOIN Three Tables

```python
# Join three related tables

cursor.execute('''
SELECT a.Title AS artist, ar.Name AS album, t.name AS track
FROM albums AS a
INNER JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId
INNER JOIN tracks AS t
ON a.AlbumId = t.AlbumId
ORDER BY ar.name
''')
# albums   --> a
# artistrs --> ar

result = cursor.fetchall()
result
```

## JOINT + Pandas

```python
df = pd.read_sql_query('''
SELECT a.Title AS artist, ar.Name AS album, t.name AS track
FROM albums AS a
INNER JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId
INNER JOIN tracks AS t
ON a.AlbumId = t.AlbumId
ORDER BY ar.name
''', conn)

df
```

## Left JOIN

```python
cursor.execute('''
SELECT a.Title, ar.Name
FROM albums AS a
LEFT JOIN artists AS ar
ON  a.ArtistId = ar.ArtistId
ORDER BY ar.name
''')
result = cursor.fetchall()
result
```

## JOIN + GROUP BY

```python
cursor.execute('''
SELECT ar.Name AS artist, COUNT(a.AlbumId) AS album_count
FROM artists AS ar
LEFT JOIN albums AS a
ON  a.ArtistId = ar.ArtistId
GROUP BY ar.ArtistId, ar.Name
ORDER BY album_count DESC
''')
result = cursor.fetchall()
result
```

```python
df = pd.read_sql_query('''
SELECT ar.Name AS artist, COUNT(a.AlbumId) AS album_count
FROM artists AS ar
LEFT JOIN albums AS a
ON  a.ArtistId = ar.ArtistId
GROUP BY ar.ArtistId, ar.Name
ORDER BY album_count DESC
''', conn)

df.head(40)
```
