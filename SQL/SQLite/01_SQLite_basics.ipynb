# SQLite Basics

## Import SQLite

```python
import sqlite3
```

## Connect to Database

```python
conn = sqlite3.connect('chinook.db')
```

## Creat Cursor

```python
cursor = conn.cursor()
```

## Execute a SQL Query

```python
# Get table names
cursor.execute("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
""")
```

## Get the Results

```python
tables = cursor.fetchall()
tables
```

## Sorting Tables

```python
for t in tables:
    print(t[0])
```

## SELECT

```python
cursor.execute("""
SELECT *
FROM artists;
""")

artists = cursor.fetchall()
artists
```

## SELECT Specific Column

```python
cursor.execute("""
SELECT name
FROM artists;
""")

artists = cursor.fetchall()
artists
```

## SELECT More than One column

```python
cursor.execute("""
SELECT name, ArtistId
FROM artists;
""")

artists = cursor.fetchall()
artists
```

## SQLite + Pandas

```python
import pandas as pd
```

```python
df = pd.read_sql_query("""
SELECT *
FROM artists;
""", conn)
df.head()
```

```python
conn.close()
```
