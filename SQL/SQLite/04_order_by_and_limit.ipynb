# ORDER BY & LIMIT

```python
import sqlite3
conn = sqlite3.connect('chinook.db')
cursor = conn.cursor()
```

## Sort Results with ORDER BY

```python
# Sort result by a column
cursor.execute("""
SELECT name, milliseconds
FROM tracks
ORDER BY milliseconds
""")  # Default --> ASC
result = cursor.fetchall()
result[:20]
```

## Ascending Order

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
ORDER BY milliseconds ASC
""")  # Default --> ASC
result = cursor.fetchall()
result[:20]
```

## Descending Order

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
ORDER BY milliseconds DESC
""") 
result = cursor.fetchall()
result[:20]
```

## LIMIT

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
LIMIT 20
""") 
result = cursor.fetchall()
result
```

## Top 10 Longest Tracks

```python
cursor.execute("""
SELECT name, milliseconds
FROM tracks
ORDER BY milliseconds DESC
LIMIT 10
""") 
result = cursor.fetchall()
result
```

## Multiple Columns

```python
# Sort by multiple columns
cursor.execute("""
SELECT name, GenreId, milliseconds
FROM tracks
ORDER BY GenreId ASC, milliseconds DESC
""") 
result = cursor.fetchall()
result
```

## LIMIT with OFFSET

```python
# Skip rows before limiting
cursor.execute("""
SELECT name, milliseconds
FROM tracks
LIMIT 10 OFFSET 10 
""") # Pass the first 10 rows and than limit!
result = cursor.fetchall()
result
```

## ORDER BY + LIMIT with Pandas

```python
# Load sorted results into a DataFrame
import pandas as pd
df = pd.read_sql_query("""
SELECT name, milliseconds
FROM tracks
ORDER BY milliseconds DESC
LIMIT 10""", conn)

df
```
