## Creating Our First Dataset

```python
data = {
    "Age": [22, 25, 28, 32, 35, 38, 41, 45, 29, 31],
    "Income": [2500, 3200, 3500, 4200, 4800, 5200, 6100, 7000, 3900, 4500],
    "Purchase": [120, 180, 210, 250, 320, 350, 420, 500, 230, 280]
}

df = pd.DataFrame(data)

df
```

## Visualization by Matplotlib

```python
plt.figure(figsize = (10, 5))

plt.hist(df['Age'], bins = 20)

plt.title('Age')
plt.xlabel('Frequency')
plt.ylabel('Age Distribution')

plt.show()
```
