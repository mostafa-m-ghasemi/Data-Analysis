# Descriptive Statistics

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Score": [65, 70, 72, 75, 80, 82, 85, 88, 90, 95]
}

df = pd.DataFrame(data)

df
```

## Mean vs Median

```python
print('Mean: ', df['Score'].mean())
print('Median: ', df['Score'].median())
```

```python
# Compare mean and median

#The mean and median are close, the dataset is not strongly skewed
```

## Mode

```python
values = [10, 20, 20, 30, 40]

# Find the mode
mode = pd.Series(values).mode()

print(mode)
```

```python
# The mode is the most frequently occurring value
```

## Min and Max

```python
print('Min: ', df['Score'].min())
print('Max: ', df['Score'].max())
```

## Range

```python
# Measure the distance between the min and max

score_range = df['Score'].max() - df['Score'].min()

print('Ranage: ', score_range)
```

## Variance

```python
# Measure how spread out data is around the mean

scores = np.array([
    65, 70, 72, 75, 80,
    82, 85, 88, 90, 95
])

# Calculate population variance
variance = np.var(scores)

print("Variance:", variance)
```

## Population vs Sample Variance

```python
# Distinguishes population from a sample
```

```python
# Calculate population variance
population_variance = np.var(scores, ddof=0)

print("Population Variance:", population_variance)
```

```python
# Calculate population variance
population_variance = np.var(scores, ddof=0)

print("Population Variance:", population_variance)
```

## Standard Diviation

```python
# Measures spread using the original data units

std = np.std(scores)
print('Standard Diviation; ', std)
```

```python
# Standard diviation measures the spread of data around the mean
# Standard diviation give us a measure of how widly values vary around the mean
```

## Peercentiles

```python
# Shows the relative position of a value in the data
```

```python
p25 = np.percentile(scores, 25)

p25
```

```python
# About 25% of the observations are blow the percentile
```

## Quartiles

```python
# Calculate quartiles
q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

```python
#### Q2 = Median
```

## IQR --> Interquartile Range

```python
# Measures the spread of the middle 50% of data
```

```python
iqr = q3 - q1

print('IQR: ', iqr)
```

## Histogram

```python
# Visulizes the distribution of numerical data
```

```python
plt.figure(figsize=(8, 5))

# Plot score distribution
plt.hist(scores, bins=5)

plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Score Distribution")

plt.show()
```

## Box Plot

```python
# Visualizees qurtiles, median, spread, and outliers
```

```python
plt.figure(figsize=(8, 4))

# Create a box plot
plt.boxplot(scores)

plt.ylabel("Score")
plt.title("Score Distribution")

plt.show()
```
