## Shape of Distribution

```python
# The Shape of Distribution describes the overall pattern of data values
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create reproducible random samples
rng = np.random.default_rng(42)

data_a = rng.normal(loc=50, scale=10, size=1000)
data_b = rng.uniform(low=20, high=80, size=1000)

# Plot the first distribution
plt.hist(data_a, bins=30, alpha=0.7, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# Plot the second distribution
plt.hist(data_b, bins=30, alpha=0.7, edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Symmetry

```python
# A Distribution is Symmetry when its left and right sides have similar shapes around the center
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate symmetric data
rng = np.random.default_rng(42)
data = rng.normal(loc=100, scale=15, size=2000)

# Calculate summary statistics
mean_value = np.mean(data)
median_value = np.median(data)

print("Mean:", round(mean_value, 2))
print("Median:", round(median_value, 2))

# Plot the histogram
plt.hist(data, bins=40, edgecolor="black")

# Mark the mean and median
plt.axvline(mean_value, linestyle="--", label="Mean")
plt.axvline(median_value, linestyle=":", label="Median")

plt.title("Symmetric Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

## Skewness

```python
# Skewness measure the asymmetry of a distribution
```

### Zero or Near-Zero Skewness

```python
# The distribution is approximately symmetriv
```

### Positive Skewness - Right-Skewness

```python
# The distribution has a longer tail on the right
```

### Positive Skewness - Left-Skewness

```python
# The distribution has a longer tail on the left
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate reproducible samples
rng = np.random.default_rng(42)

symmetric = rng.normal(loc=50, scale=10, size=3000)
right_skewed = rng.exponential(scale=10, size=3000)
left_skewed = -rng.exponential(scale=10, size=3000)

# Store samples in a DataFrame
df = pd.DataFrame({
    "Symmetric": symmetric,
    "Right Skewed": right_skewed,
    "Left Skewed": left_skewed
})

# Calculate skewness
print(df.skew().round(3))

# Plot each distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, column in zip(axes, df.columns):
    ax.hist(df[column], bins=35, edgecolor="black")
    ax.set_title(column)
    ax.set_xlabel("Value")
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### The Emprical Rule

```python
# For a normal distribution, approximately 68%, 95%, and 99.7% of values fall within one, two, and three standard diviation of the mean
```

```python
import numpy as np

# Generate normal data
rng = np.random.default_rng(42)
data = rng.normal(loc=100, scale=15, size=100_000)

# Calculate the mean and standard deviation
mean_value = np.mean(data)
std_value = np.std(data)

# Calculate percentages within standard deviation ranges
within_1 = np.mean(np.abs(data - mean_value) <= std_value) * 100
within_2 = np.mean(np.abs(data - mean_value) <= 2 * std_value) * 100
within_3 = np.mean(np.abs(data - mean_value) <= 3 * std_value) * 100

print(f"Within 1 SD: {within_1:.2f}%")
print(f"Within 2 SD: {within_2:.2f}%")
print(f"Within 3 SD: {within_3:.2f}%")
```

### Detecting Outliers with the IQR Method

```python
# The IQR metthod identifies potential outliers using the interquartile range.
```

```python
import pandas as pd

# Create sample data
data = pd.Series([2500, 2700, 2800, 3000, 3200, 3100, 2900, 50000])

# Calculate quartiles
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:")
print(outliers)
```

## Box Plot

```python
# A Box Plot summarizes a distribution using quartiles, the median, and potential outliers.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create sample data
income = pd.Series([2500, 2700, 2800, 3000, 3200, 3100, 2900, 50000])

# Create a box plot
plt.boxplot(income)

plt.title("Income Distribution")
plt.ylabel("Income")

plt.show()
```

### Compare Multiple Distributions

```python
import numpy as np
import matplotlib.pyplot as plt

# Create reproducible sample data
rng = np.random.default_rng(42)

class_a = rng.normal(70, 8, 100)
class_b = rng.normal(80, 5, 100)
class_c = rng.normal(65, 12, 100)

# Compare the distributions
plt.boxplot([class_a, class_b, class_c], tick_labels=["Class A", "Class B", "Class C"])

plt.title("Score Comparison Across Classes")
plt.xlabel("Class")
plt.ylabel("Score")

plt.show()
```

## Mini Project

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create a sample dataset
scores = pd.Series([
    55, 60, 62, 65, 67, 68, 70, 72,
    73, 74, 75, 76, 78, 80, 82, 85, 98
])

# Calculate descriptive statistics
mean_score = scores.mean()
median_score = scores.median()
skewness = scores.skew()
std_score = scores.std()

print("Mean:", round(mean_score, 2))
print("Median:", round(median_score, 2))
print("Standard Deviation:", round(std_score, 2))
print("Skewness:", round(skewness, 3))

# Calculate the IQR boundaries
Q1 = scores.quantile(0.25)
Q3 = scores.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = scores[
    (scores < lower_bound) |
    (scores > upper_bound)
]

print("\nPotential Outliers:")
print(outliers)

# Create a histogram
plt.hist(scores, bins=8, edgecolor="black")

plt.axvline(mean_score, linestyle="--", label="Mean")
plt.axvline(median_score, linestyle=":", label="Median")

plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()

plt.show()

# Create a box plot
plt.boxplot(scores)

plt.title("Box Plot of Exam Scores")
plt.ylabel("Score")

plt.show()
```
