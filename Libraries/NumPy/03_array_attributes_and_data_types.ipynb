# Numpy Array Attributes & Data Types

```python
import numpy as np
```

```python
arr = np.array([[10, 20, 30],[40, 50, 60]])
```

## ndim

```python
# Number of dimensions
```

```python
arr.ndim
```

```python
# 1D array
```

```python
np.array([1, 2, 3]).ndim
```

```python
# 2D array
```

```python
np.array([[1, 2],[3, 4]]).ndim
```

## shape

```python
# Size of each dimension
```

```python
arr.shape
```

```python
# Row
```

```python
arr.shape[0]
```

```python
# Columns
```

```python
arr.shape[1]
```

## size

```python
# Total number of elements
```

```python
arr.size
```

## dtype

```python
arr.dtype
```

## Common Data Types

```python
np.array([1, 2, 3]).dtype
```

```python
np.array([1.5, 2.3]).dtype
```

```python
np.array(['A','B']).dtype
```

```python
np.array([True, False]).dtype
```

## Explicit dtype

```python
np.array([1, 2, 3], dtype = np.int32)
```

```python
np.array([1, 2, 3], dtype = np.float64).dtype
```

## Type Conversion - astype

```python
arr = np.array([1, 2, 3, 4])
```

```python
# Convert integer to float
```

```python
arr2 = arr.astype(float)
arr2
```
