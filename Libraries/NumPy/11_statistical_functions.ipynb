# NumPy Statistical Functions

```python
import numpy as np
```

```python
data = np.array([10, 20, 30, 40, 50])
```

# Mean

```python
np.mean(data)
```

# Median

```python
np.median(data)
```

# Variance

```python
np.var(data)
```

# Standard deviation

```python
np.std(data)
```

# Percentile

```python
# 25th percentile
np.percentile(data, 25)
```

```python
# 50th percentile
np.percentile(data, 50)
```

```python
# 75th percentile
np.percentile(data, 75)
```

# Quantile

```python
# 25th quantile
np.quantile(data, 0.25)
```

```python
# 50th quantile
np.quantile(data, 0.5)
```

```python
# 75th quantile
np.quantile(data, 0.75)
```

```python
# percentile --> 0 to 100
# quantile --> 0 to 1
```

# Correlation

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
```

```python
# corelation matrix
np.corrcoef(x, y)
```

```python
# Corelation coefficient
np.corrcoef(x, y)[0, 1]
```

# Covariance

```python
# Covariance matrix
np.cov(x, y)
```

```python
# Covariance value
np.cov(x, y)[0, 1]
```

# Statistics Along an Axis

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80 ,90]
])
```

```python
# Mean of eaech column
np.mean(data, axis = 0)
```

```python
# Median for each column
np.median(data, axis = 0)
```

```python
# Standard diviation of each column
np.std(data, axis = 0)
```

```python
# Mean of each row
np.mean(data, axis = 1)
```

# Example - Data Analysis

```python
scores = np.array([65, 72, 80, 85, 90, 95, 100])
```

```python
# Basic statistics
print('Mean',np.mean(scores))
print('Median',np.median(scores) )
print('Std',np.std(scores) )
print('Min',np.min(scores) )
print('Max',np.max(scores) )
```

```python
# Quantiles
q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)
q1, q2, q3
```
