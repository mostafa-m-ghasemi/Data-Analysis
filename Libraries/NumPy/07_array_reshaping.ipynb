# NumPy Array Reshaping

```python
import numpy as np
```

```python
arr = np.arange(1, 13)
arr
```

```python
# Reshape to 3 rows and 4 columns
arr.reshape(3,4)
```

```python
# Reshape to 4 rows and 3 columns
arr.reshape(4, 3)
```

## Using

```python
# Automaticlly calculate the number of rows
arr.reshape(-1, 4)
```

```python
# Automaticlly calculate the number of columns
arr.reshape(3, -1)
```

## flatten()

```python
# Convert to 1D array
arr.flatten()
```

## ravel()

```python
arr.ravel()
```

```python
# flatten() --> returns a copy
# ravel() --> usually returns a view
```

## tranpose()

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
arr
```

```python
# Tranpose the array
arr.transpose()
```

```python
arr.T
```

## expand_dims()

```python
arr = np.array([1, 2, 3, 4])
arr.shape
```

```python
# Add a new axis
np.expand_dims(arr, axis = 0)
```

```python
# Add a new axis
np.expand_dims(arr, axis = 1)
```

## squeeze()

```python
arr = np.array([[1, 2, 3, 4]])
arr.shape
```

```python
# Remove dimensions of sixze 1
np.squeeze(arr).shape
```

# Example-Data Analysis

```python
data = np.arange(1, 13)
```

```python
# Convert one-dimensional data into a table
data.reshape(3, 4)
```

```python
# Conver the table back to 1D
data.reshape(3, 4).flatten()
```
