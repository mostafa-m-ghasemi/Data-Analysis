# Sorting & Searching

```python
import numpy as np
```

## Sorting

```python
x= np.array([5, 2, 8, 1, 9, 3])
np.sort(x)
```

```python
# Descending order
```

```python
np.sort(x)[::-1]
```

# Sorting 2D Arrays

```python
x= ([
    [5, 2, 8],
    [1, 9, 3]
])

np.sort(x)
```

```python
# Sort by column
```

```python
np.sort(x, axis = 0)
```

## argsore()

```python
x = np.array([50, 20, 80, 10])

idx = np.argsort(x)
idx
```

```python
x[idx]
```

```python
names = np.array(['Ali', 'Sara', 'Reza', 'Mina'])
scores = np.array([80, 95, 70, 100])
idx = np.argsort(scores)
idx
```

```python
names[idx]
```

## Unique()

```python
x = np.array([1, 2, 2, 3, 3, 3, 4])
np.unique(x)
```

```python
values, counts = np.unique(x, return_counts = True)

print(values)
print(counts)
```

## isin()

```python
x = np.array([10, 20, 30, 40, 50])
np.isin(x, [20, 40])
```

```python
x[np.isin(x, [20, 40])]
```

## nonzero()

```python
x = np.array([0, 5, 0, 8, 3])
np.nonzero(x)
```
