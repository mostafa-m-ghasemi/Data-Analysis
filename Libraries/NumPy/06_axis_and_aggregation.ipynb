# NumPy Axis & Aggregation

```python
import numpy as np
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]])
```

```python
# Understandin axis
# axis = 0 --> Down the rows -->
# calculate for each column
# axis = 1 --> Across the columns -->
# calculate for each row
```

## sum()

```python
# Sum of all values
np.sum(arr)
```

```python
# Sum of each column
np.sum(arr, axis = 0)
```

```python
# Sum of each row
np.sum(arr, axis = 1)
```

## mean()

```python
np.mean(arr)
```

```python
# Mean of each column
np.mean(arr, axis = 0)
```

```python
# Mean of each row
np.mean(arr, axis = 1)
```

## min() / max()

```python
# Minimu of each column
np.min(arr, axis = 0)
```

```python
# Maximum of each row
np.max(arr, axis = 1)
```

# argmmin() / argmax()

```python
arr = ([10, 50, 20, 80, 30])
```

```python
# Index of minimum value
np.argmin(arr)
```

```python
# Index of maximum value
np.argmax(arr)
```

```python
arr = ([
    [10, 50, 20],
    [80, 30, 60]])
```

```python
# Index of maximum value for each row
np.argmax(arr, axis = 1)
```

```python
# Index of minimum value for each column
np.argmin(arr, axis = 0)
```

## cumsum()

```python
arr = np.array([10, 20, 30, 40])
```

```python
# Cumulative sum
np.cumsum(arr)
```

## cumprod()

```python
# Cumulative prod
np.cumprod(arr)
```

## diff()

```python
arr = np.array([10, 15, 25, 20,  30])
```

```python
# Diffrence between consecutive values
np.diff(arr)
```
