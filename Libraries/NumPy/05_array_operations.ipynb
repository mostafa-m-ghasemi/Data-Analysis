# Basic Arithmetic

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
```

```python
# Addition
arr + 5
```

```python
# Subtraction
arr - 5
```

```python
# Multiplication
arr * 2
```

```python
# Division
arr / 2
```

```python
# Floor division
arr // 2
```

```python
# Modulo
arr % 3
```

```python
# Power
arr **2
```

# Element wise Operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

```python
# Element-wise additioin
a + b
```

```python
# Element-wise multiplication
a * b
```

```python
# Element-wise division
a / b
```

# Comparison Operators

```python
arr = np.array([10, 20, 30, 40, 50])
```

```python
# Greater than
arr > 25
```

```python
# Less than
arr < 25
```

```python
# Equal
arr == 30
```

```python
# Not equal
arr != 30
```

# Basic Mathematical Functions

```python
# Absolute value
np.abs([-10, -5, 0, 5, 10])
```

```python
# Power
np.power([2, 3, 4], 2)
```

```python
# Square root
np.sqrt([9, 16, 25])
```

```python
# Exponential
np.exp([1, 2, 3])
```

```python
# Natural logarithm
np.log([1, np.e, np.e**2])
```

```python
# Base-10 logarithm
np.log10([1, 10, 100, 1000])
```

# Rounding Functions

```python
arr = np.array([1.231, 2.5, 3.7, 4.9])
```

```python
# Round values
np.round(arr)
```

```python
# Round to 2 decimal placees
np.round(arr, 2)
```

```python
# Round down
np.floor(arr)
```

```python
# Round up
np.ceil(arr)
```

# Aggregation Functions

```python
arr = np.array([10, 20, 30, 40, 50])
```

```python
# # Minimum value
np.min(arr)
```

```python
# Maximum value
np.max(arr)
```

```python
# Sum of values
np.sum(arr)
```

```python
# Mean
np.mean(arr)
```

### Usefull Shortcuts

```python
print(arr.min(),
      arr.max(),
      arr.sum(),
      arr.mean()
     )
```
