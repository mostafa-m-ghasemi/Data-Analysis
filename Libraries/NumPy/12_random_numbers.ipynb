# Random Generator

```python
import numpy as np
```

```python
rng = np.random.default_rng()
```

```python
rng.random()
```

```python
# Random float between 0 and 1
rng.random(5)
```

## Seed

```python
# Same seed ==? same result
rng = np.random.default_rng(42)
rng.random(5)
```

## Random Integers

```python
rng.integers(1, 10)
```

```python
# 5 random integers
rng.integers(1, 10, size = 5)
```

## Random Floats

```python
rng.uniform(10, 20, size = 5)
```

## Normal Distribution

```python
rng.normal(loc=50, scale=10, size=5)
```

```python
# loc --> mean
# scale --> standard diviation
# size --> size
```

```python
temp = rng.normal(loc=20, scale=5, size=100)
temp
```

## Random Choice

```python
cities = np.array(['Baku', 'Tabriz', 'Tehran', 'Shiraz'])
rng.choice(cities)
```

```python
# Select 5 values
rng.choice(cities, size = 5)
```

```python
rng.choice(cities, size = 3, replace = False)
```

## Shuffle

```python
x = np.array([1, 2, 3, 4, 5])
rng.shuffle(x)
x
```

# Random Sampling

```python
data = np.arange(100)
sample = rng.choice(data, size = 10, replace = False)
sample
```

## Weighted Random Choice !!

```python
items = np.array(['A', 'B', 'C'])
probabilities = [0.1, 0.2, 0.7]
rng.choice(
    items,
    size= 10,
    p = probabilities
)
```
