# Creating Numpy Arrays

```python
import numpy as np
```

## zeros()

```python
# Creat an array filled with zeros
```

```python
np.zeros(5)
```

```python
# Creat a 2D array of zeros
```

```python
np.zeros((2,3))
```

## ones()

```python
# Creat an array filled with ones
```

```python
np.ones(5)
```

```python
# Creat a 2D array of ones
```

```python
np.ones((2, 5))
```

## full()

```python
# Creatg an array filled with a specific value
```

```python
np.full((2, 4), 7)
```

## empty()

```python
# Creat an uninitialalized array
```

```python
np.empty((2,3))
```

## arange()

```python
# Creat values from 0 to 9
```

```python
np.arange(10)
```

```python
# Start, stop, step
```

```python
np.arange(2, 10, 2)
```

```python
# Negative steps
```

```python
np.arange(10, 0, -2)
```

### important : Stop is not included

## linspace()

```python
# Creat 5 evenly spaced values between 0 and 10
```

```python
np.linspace(0, 10, 5)
```

```python
# Creat 11 evenly spaced values between 0 and 100
```

```python
np.linspace(0, 100, 11)
```

```python
# Difference :
#    arange() : specify the step
#    linespace() : specify the number of values
```

## eye()

```python
# Create an identity matrix
```

```python
np.eye(4)
```

## Specifyind dtype

```python
# Creat an integer array
```

```python
np.zeros(5, dtype = int)
```

```python
# Creat a float array
```

```python
np.ones(5, dtype = float)
```

```python
# Creat a boolean array
```

```python
np.zeros(5, dtype = bool)
```
