## Matrix Multiplication

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
A @ B
```

```python
np.matmul(A, B)
```

```python
np.dot(A, B)
```

```python
# row[0, 0] * column[0, 0] + row[0, 0] * column[0, 1]
```
