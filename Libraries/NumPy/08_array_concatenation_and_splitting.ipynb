# Numpy Array Concatenation & Splitting

```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b= np.array([4, 5,  6])
```

```python
# Join arrays
np.concatenate([a, b])
```

## 2D Arrays

```python
a = np.array([
    [1, 2],
    [3, 4]
])
b= np.array([
    [5, 6],
    [7, 8]
])
```

```python
# Concatenate rows
np.concatenate([a, b], axis = 0)
```

```python
# Concatenate columns
np.concatenate([a, b], axis = 1)
```

## vstack() / hstack()

```python
np.vstack([a, b])
```

```python
np.hstack([a, b])
```

## stack()

```python
a = np.array([1, 2, 3])
b = np.array([4, 5,  6])
```

```python
# Stavk along a new axis
np.stack([a, b], axis = 0)
```

```python
# Stavk along a new axis
np.stack([a, b], axis = 1)
```

```python
# Check the new shape
np.stack([a, b], axis = 0).shape
```

# Splitting Arrays

## split()

```python
arr =np.array([1, 2 ,3, 4, 5, 6])
```

```python
# Split into 3 equal parts
np.split(arr, 3)
```

```python
# Split into 2 equal parts
np.split(arr, 2)
```

## array_split()

```python
arr =np.array([1, 2 ,3, 4, 5])
```

```python
# Split into 2  parts
np.array_split(arr, 2)
```

# 2D Splitting

```python
arr = np.arange(1, 13).reshape(3, 4)
arr
```

```python
# Split rows
np.split(arr, 3, axis = 0)
```

```python
# Split columns
np.split(arr, 2, axis = 1)
```
