```python
# NumPy Quick Review
# Data Analysis Cheat Sheet

import numpy as np


# 01. NumPy Basics
# Create arrays
x = np.array([1, 2, 3])

x.ndim       # Dimensions
x.shape      # Shape
x.size       # Number of elements
x.dtype      # Data type


# 02. Creating Arrays
# Common constructors
np.zeros(5)                  # Zeros
np.ones(5)                   # Ones
np.full(5, 7)                # Filled array
np.arange(0, 10, 2)          # Range by step
np.linspace(0, 10, 5)        # Range by count
np.eye(3)                    # Identity matrix


# 03. Attributes & Data Types
# Type handling
x.dtype
x.astype(float)              # Change type


# 04. Indexing & Slicing
# Select elements
x[0]                         # Index
x[1:3]                       # Slice
x[x > 1]                     # Boolean filter
np.where(x > 1)              # Find indexes


# 05. Array Operations
# Element-wise operations
x + 2
x - 2
x * 2
x / 2
x ** 2

np.abs(x)                    # Absolute
np.sqrt(x)                   # Square root
np.log(x)                    # Natural log
np.floor(x)                  # Floor
np.ceil(x)                   # Ceiling

x.min()                      # Minimum
x.max()                      # Maximum
x.sum()                      # Sum
x.mean()                     # Mean


# 06. Axis & Aggregation
# Aggregate by axis
x2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

x2.sum(axis=0)               # Per column
x2.sum(axis=1)               # Per row
x2.mean(axis=0)              # Column mean
x2.argmax()                  # Max index
np.cumsum(x)                 # Cumulative sum
np.diff(x)                   # Differences


# 07. Reshaping
# Change array shape
x2.reshape(3, 2)
x2.flatten()                 # Copy
x2.ravel()                   # View if possible
x2.T                         # Transpose
np.expand_dims(x, axis=0)    # Add dimension
np.squeeze(x2)               # Remove size-1 dimensions


# 08. Concatenation & Splitting
# Combine and split arrays
a = np.array([1, 2])
b = np.array([3, 4])

np.concatenate([a, b])       # Combine
np.stack([a, b])             # New axis
np.vstack([a, b])            # Vertical
np.hstack([a, b])            # Horizontal
np.array_split(a, 2)         # Split


# 09. Broadcasting
# Apply operations across shapes
x + 10

A = np.ones((3, 4))
B = np.ones((1, 4))
A + B                         # Compatible shapes


# 10. Missing & Special Values
# Handle NaN and infinity
data = np.array([1, 2, np.nan, 4])

np.isnan(data)                # Find NaN
np.nanmean(data)              # Mean without NaN
np.nansum(data)               # Sum without NaN
np.isfinite(data)             # Check finite
np.isinf(data)                # Check infinity
np.nan_to_num(data)           # Replace special values


# 11. Statistical Functions
# Measure distribution
x = np.array([10, 20, 30, 40, 50])

x.mean()                      # Mean
np.median(x)                  # Median
x.var()                       # Variance
x.std()                       # Standard deviation
np.percentile(x, 75)          # Percentile
np.quantile(x, 0.75)           # Quantile
np.cov(x, x)                  # Covariance
np.corrcoef(x, x)             # Pearson correlation


# 12. Random Numbers
# Generate random data
rng = np.random.default_rng(42)

rng.random(5)                 # 0–1 floats
rng.integers(1, 10, 5)        # Random integers
rng.uniform(10, 20, 5)        # Uniform floats
rng.normal(50, 10, 5)         # Normal distribution
rng.choice(x, size=3)         # Random choice
rng.shuffle(x)                # Shuffle in-place

# Weighted choice
items = np.array(["A", "B", "C"])
p = [0.1, 0.2, 0.7]
rng.choice(items, size=10, p=p)  # Weighted choice


# 13. Sorting & Searching
# Sort and search
x = np.array([50, 20, 80, 10])

np.sort(x)                    # Sort
np.argsort(x)                 # Sorted indexes
np.searchsorted(np.sort(x), 30) # Insert position
np.unique(x)                  # Unique values
np.isin(x, [20, 50])          # Check membership
np.nonzero(x)                 # Non-zero indexes


# 15. NumPy in Data Analysis
# Basic analysis workflow
data = np.array([10, 15, 20, 25, 30, 100], dtype=float)

mask = data > 50              # Find logical outliers
data[mask] = np.nan            # Replace with NaN

np.nanmean(data)              # Mean
np.nanmedian(data)            # Median
np.nanmin(data)                # Minimum
np.nanmax(data)                # Maximum

normalized = (
    (data - np.nanmin(data)) /
    (np.nanmax(data) - np.nanmin(data))
)                             # Normalize to 0–1

standardized = (
    (data - np.nanmean(data)) /
    np.nanstd(data)
)                             # Standardize
```
