# NumPy Indexing & Slicing

## 1D Array - Indexing

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])
```

```python
# First element - Last element - Third element
```

```python
arr[0]
```

```python
arr[-1]
```

```python
arr[2]
```

```python
# Changing value
```

```python
arr[1] = 25
arr
```

## 1D Array Slicing

```python
# arr[start:stop]
arr[1:4]
```

```python
# From the begining
arr[:3]
```

```python
# To the end
arr[2:]
```

```python
# Every Second element
arr[::2]
```

```python
# Revers the array
arr[::-1]
```

## 2d Array Indexing

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
```

```python
# First row
arr[0]
```

```python
# First element of first row
arr[0, 0]
```

```python
# Row 2 column 3
arr[1,2]
```

```python
# Last element
arr[-1, -1]
```

## Selecting Rows & Columns

```python
# First , Second and last row
```

```python
arr[0,:]
```

```python
arr[1, :]
```

```python
arr[-1, :]
```

```python
# First, Second and last column
```

```python
arr[: , 0]
```

```python
arr[:,1]
```

```python
arr[:, -1]
```

## 2D Slicing

```python
# Row 1-2, column 1-2
arr[0:2,0:2]
```

```python
# First 2 rows
arr[:2,:]
```

```python
# Last two columns
arr[:, -2:]
```

```python
# Every second row
arr[::2,:]
```

```python
# Every second column
arr[:,::2]
```

# Boolean Indexing

```python
arr = np.array([10, 25, 30, 15, 40, 5])
```

```python
# Creat a boolean mask
arr > 20
```

```python
# Select values greater than 20
arr[arr > 20]
```

## Multiple Conditions

```python
# AND condition
```

```python
arr[(arr > 10) & (arr < 30)]
```

```python
# NOT condition
```

```python
arr[~(arr > 20)]
```

```python
# OR condition
```

```python
arr[(arr < 10) | (arr > 30)]
```

## np.where()

```python
arr = np.array([10, 25, 30, 15, 40])
```

```python
# Return indices where condition is True
np.where(arr>20)
```

```python
# Replace values based on a condition
np.where(arr>20, 1, 0)
```

```python
# condition --> True value --> False value
```
