# NumPy Broadcasting

## Basic Broadcasting

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40])
```

```python
# # Add a scaler to all elements
arr + 5
```

```python
# Multiple all elements
arr *2
```

## Array + Array

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

```python
a + b
```

```python
# Element-wise multipilication
a * b
```

## 2D Array + Scaler

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
arr + 10
```

```python
# Multiple all elemnets 
arr *2
```

## 2D Array + 1D Array

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
values = np.array([1, 2, 3])
arr +values
```

```python
values = np.array([
    [10],
    [20]
])

arr + values
```
