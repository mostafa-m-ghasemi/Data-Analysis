# NumPy Missing Values & Special Values

## NaN

```python
import numpy as np
arr = np.array([10, 20, np.nan, 40, 50])
arr
```

```python
# Check for NaN
np.isnan(arr)
```

```python
# Select NaN values
arr[np.isnan(arr)]
```

```python
# Select non-NaN values
arr[~np.isnan(arr)]
```

## NaN and Calculations

```python
# NaN affect calculates
np.mean(arr)
```

```python
# Mean ignoring NaN
np.nanmean(arr)
```

```python
# Sum ignoring NaN
np.nansum(arr)
```

```python
# Minimum ignoring NaN
np.nanmin(arr)
```

```python
# Maximum ignoring NaN
np.nanmax(arr)
```

## isnan() + Filtering

```python
arr = np.array([10, 20, np.nan, 30, np.nan, 40])
```

```python
# Remove NaN values
clean_arr = arr[~np.isnan(arr)]
clean_arr
```

# Replacing NaN

```python
arr = np.array([10, 20, np.nan, 40, 50])
```

```python
# replace NaN with 0
np.nan_to_num(arr, nan = 0)
```

## None

```python
arr = np.array([10, 20, None, 40, 50])
```

```python
# Check dtype
arr.dtype
```

```python
# Check for None
arr == None
```

# Example - Data Analysis

```python
temp = np.array([20.5, 22.1, np.nan, 18.7, np.nan, 25.3])
```

```python
# Count missing values
np.isnan(temp).sum()
```

```python
# Mean without missing values
np.nanmean(temp)
```

```python
# Remove missing values
temp[~np.isnan(temp)]
```
