# Indexing & Selection

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    'Name': ['Ali','Sara','Reza'],
    'Age': [25, 30, 22],
    'Score': [85, 92, 78]
})
```

## Select One Column

```python
df['Age']
```

```python
df.Age
```

## Select Multiple Columns

```python
df[['Name', 'Age']]
```

## Select Rows with iloc

```python
# Select first row
df.iloc[0]
```

```python
# Select rows 0 to 2
df.iloc[0:3]
```

```python
# Select specific rows
df.iloc[[0, 2]]
```

# Select Rows and Columns with iloc

```python
# Rows 0 to 2, columns 0 and 1
df.iloc[0:2, 0:2]
```

```python
# Specific rows and columns
df.iloc[[0, 2],[1, 2]]
```

## Select with loc

```python
# Select row by label
df.loc[0]
```

```python
# Select rows
df.loc[0:2]
```

```python
# Select rows and columns
df.loc[0:2, ['Name', 'Age']]
```

## iloc vs loc

```python
# Position-based
df.iloc[0,1]
```

```python
# Label-based
df.loc[0,'Age']
```

## Boolean Indexing

```python
# Creat a condition
condition = df['Age'] > 25
condition
```

```python
# Filter rows
df[condition]
```

```python
df[df['Age'] > 25]
```

## Multiple Conditions

```python
# AND
df[(df['Age'] > 25) & (df['Score'] > 80)]
```

```python
# OR
df[(df['Age'] > 25) | (df['Score'] < 80)]
```

```python
# NOT
df[~(df['Age'] > 25)]
```

## isin()

```python
# Select specific values
df[df['Name'].isin(['Ali', 'Sara'])]
```

## between()

```python
df[df['Age'].between(20, 30)]
```

## Select Specific Columns After Filtering

```python
# df.loc[condition, columns]
```

```python
df.loc[df['Age'] > 25, ['Name', 'Score' ]]
```

## Chane a Value with loc

```python
df.loc[df['Name'] == 'Ali', 'Score'] = 90
df
```

## Change Multiple Values

```python
df.loc[df['Score'] < 80, 'Passed'] = False
df
```

## Reset Index

```python
df = df.reset_index(drop = True)
df.to_pickle('df03.pkl')
df
```
