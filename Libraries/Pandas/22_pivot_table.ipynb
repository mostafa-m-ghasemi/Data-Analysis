# Pivot Table

```python
import pandas as pd

sales = pd.DataFrame({
    "city": ["Baku", "Baku", "Baku", "Tehran", "Tehran", "Tehran"],
    "product": ["Laptop", "Mouse", "Laptop", "Laptop", "Mouse", "Mouse"],
    "quantity": [2, 10, 3, 4, 15, 8],
    "revenue": [2400, 300, 3600, 4800, 450, 240]
})

sales
```

## Basic pivot_table()

```python
pivot = pd.pivot_table(
    sales,
    values = 'revenue',
    index = 'city',
    columns = 'product',
    aggfunc = 'sum'
)
pivot
```

## Using mean()

```python
pivot = pd.pivot_table(
    sales,
    values = 'revenue',
    index = 'city',
    columns = 'product',
    aggfunc = 'mean'
)
pivot
```

## Using Multiple Aggregations

```python
pivot = pd.pivot_table(
    sales,
    values = 'revenue',
    index = 'city',
    columns = 'product',
    aggfunc = ['min', 'max', 'mean']
)
pivot
```

## Multiple Values

```python
pivot = pd.pivot_table(
    sales,
    values = ['revenue', 'quantity'], 
    index = 'city',
    columns = 'product',
    aggfunc = 'sum'
)
pivot
```

## Adding Total with margins

```python
# Add grand totals

pivot = pd.pivot_table(
    sales,
    values = 'revenue',
    index = 'city',
    columns = 'product',
    aggfunc = 'sum',
    margins = True
)
pivot
```

## fill_value

```python
# Replace missing combinations with 'sth'

pivot = pd.pivot_table(
    sales,
    values = 'revenue',
    index = 'city',
    columns = 'product',
    aggfunc = 'sum',
    fill_value = 0
)
pivot
```

## Practical Analysis

```python
sales = pd.DataFrame({
    "city": ["Baku", "Baku", "Baku", "Tehran", "Tehran", "Tehran"],
    "product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse", "Keyboard"],
    "revenue": [2400, 300, 400, 4800, 450, 600]
})

sales
```

```python
report = pd.pivot_table(
    sales,
    values="revenue",
    index="city",
    columns="product",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

report
```
