# Pandas Basics

```python
# import pandas
import pandas as pd
```

# Series

```python
# Creat a Series
s = pd.Series([10, 20, 30, 40, 50])
s
```

```python
# Custom index
s = pd.Series(
    [1, 2, 3],
    index = ['A','B','C'])
s
```

```python
# Accws by labe;
s['B']
```

```python
# Acces by position
s.iloc[1]
```

# DataFrame

```python
# Creat a DataFrame
df = pd.DataFrame({
    'Name': ['Ali','Sara','Reza'],
    'Age': [25, 30, 22],
    'Score': [85, 92, 78]
})
df
```

```python
# Show first rows
df.head(2)
```

```python
# Show last rows
df.tail(2)
```

```python
# Data type
df.dtypes
```

# Selecting Columns

```python
# select one column
df['Age']
```

```python
# select multiple columns
df[['Name','Score']]
```

# Selecting Rows

```python
# Select by position
df.iloc[0]
```

```python
# Select multiple rows
df.iloc[0:2]
```

```python
# Select specific rowa and columns
df.iloc[0:2,0:2]
```

# Basic Filtering

# Filter rows
df[df['Age'] > 24]

```python
# Multiple condition
df[(df['Age'] > 24)&(df['Score'] > 85)]
```

```python
# OR condition
df[(df['Age'] > 24)|(df['Score'] > 90)]
```

# Adding a column

```python
# creat a new column
df['Passed'] = df['Score'] >= 80
df
```
