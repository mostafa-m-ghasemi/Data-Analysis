# Pivot Table and Crosstab

```python
import pandas as pd
```

## Creat DataFrame

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "Sales", "Sales", "HR", "HR", "IT", "Sales"],
    "Gender": ["Male", "Female", "Male", "Female", "Female", "Male", "Male", "Female"],
    "Salary": [3000, 3500, 2800, 3200, 2500, 2700, 3300, 3100],
    "Experience": [2, 4, 3, 5, 2, 6, 4, 3]
})

df
```

## Basic Pivot Table

```python
pd.pivot_table(
    df,
    values = 'Salary',
    index = 'Department',
    aggfunc = 'mean'
)
```

## Multiple Aggregations

```python
pd.pivot_table(
    df,
    values = 'Salary',
    index = 'Department',
    aggfunc = ['mean', 'max', 'min']
)
```

## Add Columns

```python
pd.pivot_table(
    df,
    values = 'Salary',
    columns = 'Gender',
    index = 'Department',
    aggfunc = 'mean'
)
```

## Multiple Values

```python
pd.pivot_table(
    df,
    values = ['Salary', 'Experience'],
    index = 'Department',
    aggfunc = 'mean'
)
```

## Fill Missing Values

```python
pd.pivot_table(
    df,
    values = ['Salary', 'Experience'],
    columns = 'Gender',
    index = 'Department',
    aggfunc = 'mean',
    fill_value = 0
)
```

## Crosstab

```python
pd.crosstab(
    df['Department'],
    df['Gender']
)
```

## Crosstab wuth Percentage

```python
pd.crosstab(
    df['Department'],
    df['Gender'],
    normalize = 'index'
) *100
```

## Pivot Table with Multiple Indexes

```python
pd.pivot_table(
    df,
    values = 'Salary',
    index = ['Department', 'Gender'],
    aggfunc = 'mean'
)
```
