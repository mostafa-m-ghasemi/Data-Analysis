# Data Types & Conversion

```python
import numpy as np
import pandas as pd
df = pd.read_pickle('df03.pkl')
df
```

## Check Data Types

```python
# Check data types
df.dtypes
```

```python
# Check one column type
df["Age"].dtype
```

## Common Pandas Data Types

```python
# int64    -->  Integer
# float64  -->  Decimal
# object   --> Text / mixed
# string   --> Text
# datatime --> Data and time
```

## Convert with astype()

```python
# Convert to integer
df['Age'] = df['Age'].astype(int)
df['Age'].dtype
```

```python
# Convert to float
df['Age'] = df['Age'].astype(float)
df['Age'].dtype
```

```python
# Convert to string
df['Age'] = df['Age'].astype(str)
df['Age'].dtype
```

```python
# Convert to bpplean
df['Age'] = df['Age'].astype(bool)
df['Age'].dtype
```

## Convert Multiple Columns

```python
df = df.astype({
    'Age': 'int64',
    'Score': 'float64'
})
```

## pd.to_numeric()

```python
# Convert to numeric
df['Age'] = pd.to_numeric(df['Age'])
df
```

```python
# Conveert invalid values to NaN
# Convert to numeric
df['Age'] = pd.to_numeric(df['Age'], errors = 'coerce')
df
```

## Convert to DateTime

```python
df['Date'] = ['2025-01-15', '2024-02-20', '2025-04-25']
df['Date'] = pd.to_datetime(df['Date'])
df
```

```python
# Convert invalid dates to NaN
df['Date'] = pd.to_datetime(df['Date'], errors = 'coerce')
df
```

## Extract Data Components

```python
# Extract year
df['Year'] = df['Date'].dt.year
df
```

```python
# Extract month
df['Month'] = df['Date'].dt.month
df
```

```python
# Extract day
df['Day'] = df['Date'].dt.day
df
```

```python
# Extract weekday
df['Weekday'] = df['Date'].dt.day_name()
df
```

```python
# Extract hour
df['Hour'] = df['Date'].dt.hour
df
```

## String Data

```python
# Convert text to lowercase
df['Name'] = df['Name'].str.lower()
df
```

```python
# Convert text to Uppercase
df['Name'] = df['Name'].str.upper()
df
```

```python
# Remove extra spaces
df['Name'] = df['Name'].str.strip()
df
```

## String Contains

```python
# Check if text contains 'ALI'
df['Name'].str.contains('ALI')
```

```python
# Filter rows containing 'ALI'
df[df['Name'].str.contains('ALI')]
```

```python
# Treat missing values as False
df[df['Name'].str.contains('ALI', na = False)]
```

## Replace text

```python
df['Name'] = df['Name'].str.replace('ALI','Ali')
df['Name'] = df['Name'].str.replace('SARA','Sara')
df['Name'] = df['Name'].str.replace('REZA','Reza')
df
```

```python
# Convert to category --> reducing memory useage
df['Name'] = df['Name'].astype('category')
```
