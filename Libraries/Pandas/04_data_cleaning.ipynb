# Data Cleaning

```python
import numpy as np
import pandas as pd
```

## Check Missing Values

```python
df = pd.read_pickle('df03.pkl')
df
```

```python
# Check missing values
df.isna()
```

```python
df.isna().sum()
```

## Check Missing Values as Percentage

```python
# Missing percentage
df.isna().mean() * 100
```

```python
# Sort missing percentage
df.isna().mean().sort_values(ascending = False) * 100
```

## Check if Missing Values Exist

```python
# Check in columns
df.isna().any()
```

```python
# Check in whole DataFrame
df.isna().any().any()
```

## Select Rows with Missing Values

```python
# Select rows with missing Passed
df[df['Passed'].isna()]
```

```python
# Select rows without missing Passed
df[df['Passed'].notna()]
```

```python
# Remove rows with any missing value
df1 = df.copy()
df1.dropna()
# Apply changes
df1.dropna(inplace = True)
```

```python
df1
```

```python
df1 = df.copy()
```

```python
# Remove rows missing Passed values
df.dropna(subset = ['Passed', 'Score'])
```

## Drop Columns with Missing Values

```python
df.dropna(axis = 1)
```

```python
# axis = 0 --> rows
# axis = 1 --> columns
```

## Fill Missing Values

```python
# Fill missing values by 0
df1['Passed'] = df['Passed'].fillna(0)
```

```python
df1
```

## Fill with Mean

```python
df1 = df.copy()
```

```python
df1['Passed'] = df['Passed'].fillna(df['Passed'].mean())
```

```python
df1
```

## Fill with Median

```python
df1['Passed'] = df['Passed'].fillna(df['Passed'].median())
```

## Fill with Mode

```python
df1['Passed'] = df['Passed'].fillna(df['Passed'].mode())
```

## Forward Fill

```python
# Fill using previuos value
df1['Passed'] = df['Passed'].ffill()
```

```python
# 10 - 20 - NAN - 40 --> 10 - 20 - 20 - 40
```

## Backward Fill

```python
# Fill using next value
df1['Passed'] = df['Passed'].bfill()
```

```python
# 10 - 20 - NAN - 40 --> 10 - 20 - 40 - 40
```

## eplace Specific Values

```python
# Replace specific values
df['Age'] = df['Age'].replace(0, np.nan)
```

```python
# Replace text value
df['Name'] = df['Name'].replace('Unknown', np.nan)
```

## Remove Duplicate Rows

```python
# Check duplicated rows
df.duplicated()
```

```python
# Count duplicate rws
df.duplicated().sum()
```

```python
# Remove duplicated rows
df = df.drop_duplicates()
```

## Duplicate Based on Specific Column

```python
# Check duplicates based on Name
df.duplicated(subset = ['Name'])
```

```python
# Remove duplicated names
df = df.drop_duplicates(subset = ['Name'])
```
