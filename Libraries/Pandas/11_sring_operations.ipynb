# Sting Operations

## Creat DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali Ahmadi", "Sara Mohammadi", "Reza Karimi", "Nima Hosseini"],
    "City": ["tehran", "TEHRAN", "Mashhad", "tehran"],
    "Email": [
        "ali@gmail.com",
        "sara@yahoo.com",
        "reza@gmail.com",
        "nima@gmail.com"
    ]
})

df
```

## Conveert to Lowercase

```python
df['City'].str.lower()
```

## Convert to Uppercase

```python
df['Name'].str.upper()
```

## String Lenght

```python
df['Name'].str.len()
```

## Contains

```python
df['Email'].str.contains('gmail')
```

```python
df[df['Email'].str.contains('gmail')]
```

## Replace

```python
df['City'] = df['City'].str.replace('tehran', 'Tehran')
df
```

## Split

```python
df[['First_Name', 'Last_name']] = df['Name'].str.split(' ', expand = True)
df
```

## Extract

```python
df['Email'].str.extract(r'@(.+)')
```
