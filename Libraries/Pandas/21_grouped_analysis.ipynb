# groupby() and Grouped Analysis

```python
import pandas as pd

df = pd.DataFrame({
    "department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "employee": ["Ali", "Reza", "Sara", "Mina", "John", "David"],
    "salary": [5000, 6000, 4500, 5200, 4000, 5500],
    "age": [25, 30, 28, 35, 24, 32]
})

df
```

## Basic groupby()

```python
df.groupby('department')
```

```python
# Calculate average salary per department

df.groupby('department')['salary'].mean()
```

## sum()

```python
# Calculate total salary per department

df.groupby('department')['salary'].sum()
```

## Count()

```python
df.groupby('department')['employee'].count()
```

## min() & max()

```python
# Find the minimum salary per department

df.groupby('department')['salary'].min()
```

```python
# Find the maximum salary per department

df.groupby('department')['salary'].max()
```

## Multiple Aggregations

```python
df.groupby('department')['salary'].agg(['min', 'max', 'count', 'mean', 'sum'])
```

## Grouping Multiple Columns

```python
df.groupby('department')[['salary', 'age']].mean()
```

## Group by Multiple Columns

```python
sales = pd.DataFrame({
    "city": ["Baku", "Baku", "Tehran", "Tehran", "Baku", "Tehran"],
    "product": ["Laptop", "Mouse", "Laptop", "Mouse", "Laptop", "Laptop"],
    "sales": [1200, 300, 1500, 400, 1800, 1300]
})

sales
```

```python
sales.groupby(['city', 'product'])['sales'].sum()
```

## reset_index()

```python
result = df.groupby('department')['salary'].mean()
result
```

```python
result = result.reset_index()

result
```

## Named Aggregations

```python
result = df.groupby('department').agg(avg_salary = ('salary', 'mean'),
                                     max_salary = ('salary', 'max'),
                                     min_salary = ('salary', 'min'),
                                     employee_count = ('employee', 'count'))

result
```

## Sorting Grouped Results

```python
result = (
    df.groupby('department')['salary'].mean().sort_values(ascending = False)
)

result
```
