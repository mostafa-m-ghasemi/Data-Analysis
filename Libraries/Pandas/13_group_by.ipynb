# Group by Aggregation

## Create DataFrame

```python
import pandas as pd
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'Sales', 'Sales', 'HR', 'HR'],
    'Employees': ['Ali', 'Sara', 'Reza', 'Nima', 'Mina', 'Omid'],
    'Salary': [3000, 3500, 2800, 3200, 2500, 2700],
    'Experience': [2, 4, 3, 5, 2, 6]
})
df
```

## Group by One Column

```python
df.groupby('Department')
```

```python
# average salary 
df.groupby('Department')['Salary'].mean()
```

```python
# Total salary
df.groupby('Department')['Salary'].sum()
```

## Multiple Aggregations

```python
df.groupby('Department')['Salary'].agg(['mean', 'max', 'min', 'sum', 'count'])
```

## Group Multiple Columns

```python
df.groupby(['Department', 'Experience'])['Salary'].mean()
```

## Aggregation on Multiple Columns

```python
df.groupby('Department').agg({
    'Salary': ['min', 'max'],
'Experience': ['mean', 'count']})
```

## Named Aggregation

```python
df.groupby('Department').agg(Avg = ('Salary', 'mean'),
                            maxi = ('Salary', 'max'),
                            mini = ('Salary', 'min'),
                            sumi = ('Salary', 'sum'))
```

## GroupBy + Sort

```python
result = (df.groupby('Department')['Salary'].mean().sort_values(ascending = False))
result
```
