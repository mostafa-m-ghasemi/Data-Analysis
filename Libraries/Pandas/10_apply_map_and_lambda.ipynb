# Apply Map and Lambda

## Creat DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Reza", "Mina", "Omid"],
    "Age": [25, 30, 22, 28, 35],
    "Score": [85, 92, 78, 88, 95],
    "City": ["Baku", "Tehran", "Baku", "Tabriz", "Tehran"]
})

df
```

```python
square = lambda x: x**2
square(5)
```

```python
add_10 = lambda x : x+ 10
add_10(10)
```

## apply() on a Series

```python
df['New_Score'] = df['Score'].apply(lambda x : x+ 5)
df
```

## apply() with Function

```python
def level(score):
    if score >= 90:
        return 'Execllent'
    elif score >= 80:
        return 'Good'
    else:
        return 'Average'

df['Level'] = df['Score'].apply(level)


df
```

```python
def age(age):
    if age < 25:
        return 'Young'
    elif age < 35:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(age)

df
```

## map()

```python
df['City'].map(lambda x : x.upper())
```

```python
df['Name'].map(lambda x : x.upper())
```

```python
df['Name_upper'] = df['Name'].map(lambda x : x.upper())
df
```

## map() with Dictionary

```python
city = {
    'Baku': 'BAK',
    'Tehran': 'THR',
    'Tabriz': 'TAB'
}

df['City_Code'] = df['City'].map(city)

df
```

## apply() Multiple Cplumns

```python
df[['Age', 'Score']].apply(lambda x : x.mean())
```

## apply() with axis

```python
df[['Age', 'Score']].apply(lambda x : x.mean(), axis = 1)
```

## Creat a New Column with Multiple Columns

```python
df['Final'] = df.apply(
    lambda row: row['Score'] * .08 + row['Age'] * .02, axis = 1
)
df
```
