# Dorting and Ranking

## Creat a DataFrame

```python
import pandas as pd
df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Reza', 'Mina', 'Omid'],
    'Age': [25, 30, 22, 28, 25],
    'Score': [85, 92, 78, 92, 88],
    'City': ['Baku', 'Tehran', 'Baku', 'Tabriz', 'Tehran']
})
df
```

## Sorting by One Column

```python
# Sort by Age
df.sort_values('Age')
```

## Descending Order

```python
df.sort_values('Age', ascending = False)
```

## Sorting bu Multiple Columns

```python
## Sort by Age, than Score
df.sort_values(['Age', 'Score'])
```

## Different Order for Each Column

```python
df.sort_values(['Age', 'Score'], ascending = [True, False])
```

```python
# Age   --> Ascending
# Score --> Descending
```

## Inplace

```python
df.sort_values('Score', ascending = False, inplace = True)
df
```

## Sort by Index

```python
# Sort by index
df.sort_index()
```

```python
# Sort by index descending
df.sort_index(ascending = False)
```

## Reset Index After Sorting

```python
df.reset_index(drop = True)
```

## Ranking

```python
# Creat ranks
df['Rank'] = df['Score'].rank()
df
```

## Descending Rank

```python
df['Rank'] = df['Score'].rank(ascending = False)
df
```

## Ranking with method

```python
# Average rank for ties
df['Rank'] = df['Score'].rank(ascending = False, method = 'average')
df
```

```python
# min / max rank for ties
df['Rank'] = df['Score'].rank(ascending = False, method = 'min')
df
```

```python
# Dense ranking
df['Rank'] = df['Score'].rank(ascending = False, method = 'dense')
df
```

## Find Top Values

```python
# Top 3 Scores
df.nlargest(3, 'Score')
```

## Find Bottom Values

```python
# Bottom 3 scores
df.nsmallest(3, 'Score')
```
