## Creat DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Ali", "Sara", "Reza", "Mina", "Omid", "Nima", "Lina"],
    "Salary": [2800, 3000, 3200, 3100, 3500, 3300, 25000]
})

df
```

## Detect Outliers with IQR

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3 -Q1

print('Q1: ', Q1)
print('Q3: ', Q3)
print('IQR: ', IQR)
```

## Calculate Lower and Upper Bound

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print('Lower Bound: ' , lower)
print('Upper Bound: ' , upper)
```

## Find Outliers

```python
outliers = df[(df['Salary'] < lower) | (df['Salary'] > upper)]
outliers
```

### Find Normal Values

```python
normal = df[(df['Salary'] >= lower) & (df['Salary'] <= upper)]
normal
```

## Remove Outliers

```python
df_clean = df[(df['Salary'] >= lower) & (df['Salary'] <= upper)]
df_clean
```

## Box Plot for Outliers

```python
import matplotlib.pyplot as plt
```

```python
plt.boxplot(df['Salary'])

plt.title('Salary')
plt.ylabel('S')

plt.show()
```

```python
plt.boxplot(df_clean['Salary'])

plt.title('Salary')
plt.ylabel('S')

plt.show()
```
