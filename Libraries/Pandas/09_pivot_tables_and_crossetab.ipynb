# Pivot Tables and Crosstab

## Creat DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Baku", "Baku", "Tehran", "Tehran", "Tabriz", "Tabriz", "Baku", "Tehran"],
    "Department": ["Sales", "IT", "Sales", "IT", "Sales", "HR", "HR", "HR"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Female", "Male"],
    "Sales": [250, 400, 300, 500, 200, 350, 450, 280],
    "Score": [85, 90, 78, 92, 75, 88, 95, 82]
})

df
```

## Basic pivot_table

```python
df.pivot_table(values = 'Sales', index = 'City', aggfunc = 'mean')
```

## Important Parameters

```python
df.pivot_table(values = 'Sales', index = 'City', columns = 'Department', aggfunc = 'mean')
```

## Different Aggregation Functions

```python
df.pivot_table(values = 'Sales', index = 'City', aggfunc = ['mean', 'sum', 'count'])
```

## Multiple Values

```python
df.pivot_table(values = ['Sales', 'Score'], aggfunc = 'mean', index = 'City')
```

## Two-Dimentional Pivot Table

```python
df.pivot_table(values = 'Sales', index = 'City', columns = 'Department', aggfunc ='sum')
```

## fill_value

```python
df.pivot_table(values = 'Sales', index = 'City', columns = 'Department', aggfunc ='sum', fill_value = 0  )
```

## multiple Indexes

```python
df.pivot_table(values = 'Sales', index = ['City', 'Gender'], aggfunc = 'max')
```

## margins

```python
df.pivot_table(values = 'Sales', index = 'City', columns = 'Department', aggfunc ='sum', fill_value = 0 , margins = True )
```

## margins_name

```python
df.pivot_table(values = 'Sales',
               index = 'City',
               columns = 'Department',
               aggfunc ='sum', fill_value = 0,
               margins = True,
              margins_name = 'Total')
```

## Crosstab --> count values

```python
pd.crosstab(df['City'], df['Department'])
```

## Croostab with Normalize

```python
pd.crosstab(df['City'], df['Department'], normalize= 'index' ) * 100
```

## Crosstab with margins

```python
pd.crosstab(df['City'], df['Department'], margins = True ) 
```

## Data Analysis Example

```python
summary = df.pivot_table(
    values = 'Sales',
    index ='City',
    columns = 'Department',
    aggfunc = 'sum',
    fill_value = 0
)
summary
```
