# Merg, Join and Concatenation

## Creat DataFrame

```python
import pandas as pd
customers = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'Name': ['Ali', 'Sara', 'Reza', 'Mina', 'Omid'],
    'City': ['Baku', 'Tehran', 'Baku', 'Tabriz', 'Tehran']
})
customers
```

```python
orders = pd.DataFrame({
    'Order_Id': [1, 2, 3, 5],
    'Customer_ID': [101, 102, 103, 105],
    'Amount': [250, 400, 150, 200]
})
orders
```

## merge()

```python
df = pd.merge(customers, orders, on = 'Customer_ID')
df
```

## Inner Join

```python
## Inner marge --> Default merge --> Customers_Id in both Tables
```

```python
df = pd.merge(customers, orders, on = 'Customer_ID', how = 'inner')
df
```

## Left Join

```python
## Save all the rows in left table
```

```python
df = pd.merge(customers, orders, on = 'Customer_ID', how = 'left')
df
```

## Right Join

```python
## Save all the rows in right table
```

```python
df = pd.merge(customers, orders, on = 'Customer_ID', how = 'right')
df
```

## Outer join

```python
## Save all the records in both tables --> NaN for not matched rows
```

```python
df = pd.merge(customers, orders, on = 'Customer_ID', how = 'outer')
df
```

## Visualizing Joint Types

```python
# INNER   --> matching A and B
# LEFT    --> A + matching B
# Right   --> B + matching A
# OUTER   --> A + B
```

## Merge on Multiple Columns

```python
sales = pd.DataFrame({
    "Year": [2025, 2025, 2026],
    "Product_ID": [1, 2, 1],
    "Sales": [500, 700, 900]
})

targets = pd.DataFrame({
    "Year": [2025, 2025, 2026],
    "Product_ID": [1, 2, 1],
    "Target": [450, 800, 850]
})
```

```python
df = pd.merge(sales, targets, on = ['Year', 'Product_ID'])
df
```

## suffixes

```python
employees = pd.DataFrame({
    "Employee_ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Reza"],
    "Score": [80, 90, 85]
})

reviews = pd.DataFrame({
    "Employee_ID": [1, 2, 3],
    "Score": [85, 88, 92]
})
```

```python
df = pd.merge(employees, reviews, on ='Employee_ID', suffixes = ('_Employees', '_Reviews'))
df
```

## concat()

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Score": [85, 92]
})

df2 = pd.DataFrame({
    "Name": ["Reza", "Mina"],
    "Score": [78, 88]
})
```

```python
df = pd.concat([df1, df2])
df
```

## Reset Index After concat()

```python
df = pd.concat([df1, df2], ignore_index = True)
df
```

## Concatenate Columns

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Reza"]
})

df2 = pd.DataFrame({
    "Score": [85, 92, 78]
})
```

```python
df = pd.concat([df1, df2], ignore_index = True, axis = 1)
df
```

## Concatenate Diffrent Columns

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Score": [85, 92]
})

df2 = pd.DataFrame({
    "Name": ["Reza", "Mina"],
    "Age": [78, 88]
})
```

```python
df = pd.concat([df1, df2], ignore_index = True)
df
```

## join()

```python
left = pd.DataFrame({
    "Name": ["Ali", "Sara", "Reza"]
}, index=[101, 102, 103])

right = pd.DataFrame({
    "Score": [85, 92, 78]
}, index=[101, 102, 103])
```

```python
df = left.join(right)
df
```

## Left Join with join()

```python
df = left.join(right, how = 'left')
df
```

## Outer Join with join()

```python
right = pd.DataFrame({
    "Score": [85, 92, 78, 95]
}, index=[101, 102, 103, 104])
```

```python
df = left.join(right, how = 'outer')
df
```

## merge() VS join() VS concat()

```python
# merge()  --> Connect using columns
# join()   --> Connect using index
# concat() --> Stack / combine DtaFrames
```

## Real Data Analysis Example

```python
# Merge customers and orders
customer_orders = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)

customer_orders
```

```python
# Calculate total spending
summary = customer_orders.groupby(
    ["Customer_ID", "Name"]
)["Amount"].sum()

summary
```

```python
# Create customer summary
summary = customer_orders.groupby(
    ["Customer_ID", "Name"],
    as_index=False
)["Amount"].sum()

summary
```

```python
# Sort by total spending
summary.sort_values(
    "Amount",
    ascending=False
)
```
