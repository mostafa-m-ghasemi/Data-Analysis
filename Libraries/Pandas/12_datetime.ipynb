# Datetime Operations

## Creat DataFrame

```python
import pandas as pd
df = pd.DataFrame({
    'Order': [101, 102, 103, 104],
    'Date': [
        '2025-01-15',
        '2025-02-20',
        '2025-03-10',
        '2025-04-25'
    ],
    'Sales': [1200, 1800, 1500, 2200]
})
df
```

## Conveert to Datetime

```python
df['Date'].dtypes
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.dtypes
```

## Extract Year, Month and Day

```python
df['Year'] = df['Date'].dt.year 
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df
```

## Extract Day of Week

```python
df['Day_Name'] = df['Date'].dt.day_name()
df
```

## Filter by Date

```python
df[(df['Date'] >= '2025-02-01') & (df['Date'] <= '2025-03-31')]
```

## Sort by Date

```python
df.sort_values('Date', ascending = False)
```

## Gruop by Month

```python
monthly_sales = df.groupby(df['Date'].dt.month)['Sales'].sum()
monthly_sales
```

## Creat a Date Range

```python
dates = pd.date_range(
    start ='2025-01-01',
    end = '2025-01-10'
)
dates
```
