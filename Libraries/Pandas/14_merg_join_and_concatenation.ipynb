# Merge, Join and Concatenation

## Creat DataFrame

```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Reza", "Mina"]
})

orders = pd.DataFrame({
    "Customer_ID": [1, 2, 2, 4],
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "Amount": [1200, 800, 50, 100]
})

customers
```

## Inner Merge

```python
merged = pd.merge(
    customers,
    orders,
    on = 'Customer_ID',
    how = 'inner'
)
merged
```

## Left Merge - Right Merge

```python
# Left merge keep all the left values
left = pd.merge(
    customers,
    orders,
    on = 'Customer_ID',
    how= 'left'
) 
left
```

```python
# right merge keep all the right values
right = pd.merge(
    customers,
    orders,
    on = 'Customer_ID',
    how= 'right'
) 
right
```

## Outer Merge

```python
# outer keep all the values in both dataframes

merged = pd.merge(
    customers,
    orders,
    on = 'Customer_ID',
    how = 'outer'
)
merged
```

## Merge with Diffrent Column Names

```python
# New Data
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Reza", "Mina"]
})

orders = pd.DataFrame({
    "Client_ID": [1, 2, 2, 4],
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "Amount": [1200, 800, 50, 100]
})

merged = pd.merge(
    customers,
    orders,
    left_on = 'Customer_ID',
    right_on = 'Client_ID',
    how = 'inner'
)
merged
```

## Concatenate Rows

```python
df1 = pd.DataFrame({
    'Name': ['Ali', 'Sara'],
    'Sales': [1000, 1500]
})

df2 = pd.DataFrame({
    'Name': ['Reza', 'Mina'],
    'Sales': [1200, 1800]
})

result = pd.concat([df1, df2], ignore_index = True)

result
```

## Concatenate Clumns

```python
result = pd.concat([df1, df2], axis = 1)

result
```
