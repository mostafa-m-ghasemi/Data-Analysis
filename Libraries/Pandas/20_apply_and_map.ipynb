# apply() & map()

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Mina"],
    "age": [25, 30, 28, 35],
    "salary": [4500, 5200, 4800, 6100]
})

df
```

## Using map()

```python
df['age'] = df['age'].map(lambda x: x + 1)

df
```

## map() with a Function

```python
def tax(salary):
    return salary * 1.1
```

```python
df['salary'] = df['salary'].map(tax)

df
```

## map() with Dictionary

```python
df2 = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Mina"],
    "gender": ["M", "F", "M", "F"]
})

df2
```

```python
gender = {
    "M": "Male",
    "F": "Female"
}

# Map codes to readable labels
df2["gender"] = df2["gender"].map(gender)

df2
```

## Using apply()

```python
def salary_level(salary):
    if salary > 5500:
        return 'High'
    else:
        return 'Low'
```

```python
df['salary_level'] = df['salary'].apply(salary_level)

df
```

## apply() with Lambda

```python
df['age_group'] = df['age'].apply(lambda x: 'Young' if x < 30 else 'Adult')

df
```

## apply() on Rows

```python
def score(row):
    return row['age'] + row['salary'] / 1000
```

```python
df['score'] = df.apply(score, axis = 1)
df
```

## axis = 0 VS axis = 1

```python
# axis = 0 --> Down   --> Columns
# axis = 1 --> Across --> Rows
```

## Practical Example

```python
sales = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "price": [1200, 25, 80, 300],
    "quantity": [2, 10, 5, 3]
})

sales
```

```python
# Calculate Total Sales
sales['total'] = sales.apply(lambda row: row['price'] * row['quantity'], axis = 1)

sales
```

```python
# OR sales['total'] = sales['price'] * sales['quantity']
```
