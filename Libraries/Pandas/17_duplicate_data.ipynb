# Duplicate Data

```python
import pandas as pd
```

## Creat DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Reza", "Ali", "Mina", "Sara"],
    "Age": [25, 30, 28, 25, 32, 30],
    "City": ["Tehran", "Mashhad", "Tabriz", "Tehran", "Shiraz", "Mashhad"],
    "Salary": [3000, 3500, 2800, 3000, 4000, 3500]
})

df
```

## Detect Duplicates

```python
df.duplicated()
```

## Count Duplicates

```python
df.duplicated().sum()
```

## Show Duplicate Rows

```python
df[df.duplicated()]
```

## Remove Duplicates

```python
df_clean = df.drop_duplicates()
df_clean
```

## Duplicate Based on Specific Columns

```python
df[df.duplicated('Name', keep = False)]
```

## Keep First or Last

```python
df_clean = df.drop_duplicates(
    subset = ['Name'],
    keep = 'first'
)

df_clean
```

```python
df_clean = df.drop_duplicates(
    subset = ['Name'],
    keep = 'last'
)

df_clean
```

## Remove Duplicates Based on Multiple Columns

```python
df_clean = df.drop_duplicates(subset = ['Name', 'Age'])
df_clean
```
