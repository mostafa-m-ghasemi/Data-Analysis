## Checking Data Types

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Mina"],
    "age": ["25", "30", "28", "35"],
    "salary": ["4500.5", "5200.0", "4800.75", "6100.25"],
    "active": [1, 0, 1, 1]
})

df
```

```python
df.dtypes
```

## Converting to integer

```python
df['age'] = df['age'].astype(int)

df.dtypes
```

## Converting to float

```python
df['salary'] = df['salary'].astype(float)

df.dtypes
```

## Using 'pd.to_numeric'

```python
df2 = pd.DataFrame({
    "age": ["25", "30", "unknown", "35"]
})

df2
```

```python
# Not working because of 'unknown' values
df2['age'] = df2['age'].astype(int)
```

```python
# Convert invalid values to NAN

df2['age'] = pd.to_numeric(df2['age'], errors = 'coerce')

df2
```

## Converting to string

```python
df['age'] = df['age'].astype(str)
```

## Boolean Converting

```python
df['active']
```

```python
df['active'] = df['active'].astype(bool)

df['active']
```

## Converting Multiple Columns

```python
df = df.astype({
    'age': int,
    'salary': float
})

df
```

## Numeriv Conversion to Real Data

```python
df3 = pd.DataFrame({
    "salary": ["$4,500", "$5,200", "$4,800", "$6,100"]
})

df3
```

```python
# Remove currency symbols and cammas
df3['salary'] = df3['salary'].str.replace('$', '', regex = False)
df3['salary'] = df3['salary'].str.replace(',', '', regex = False)

df3['salary'] = pd.to_numeric(df3['salary'])

df3
```

## Category Data Type

```python
df4 = pd.DataFrame({
    "city": ["Baku", "Tehran", "Baku", "Baku", "Tehran"]
})

df4
```

```python
# Convert repeated values to category

df4["city"] = df4["city"].astype("category")

df4.dtypes
```
