# Groupby and Aggregation

```python
import pandas as pd

# Creat a DataFrame
df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Reza', 'Mina', 'Omid', 'Neda', 'Amir', 'Maryam'],
    'City': ['Baku', 'Tehran', 'Baku', 'Tehran', 'Baku', 'Tabriz', 'Tehran', 'Tabriz'],
    'Department': ['Sales', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'IT', 'HR'],
    'Age': [25, 30, 22, 28, 25, 32, 27, 29],
    'Salary': [2500, 4200, 2800, 3500, 4000, 3000, 4500, 3200],
    'Score': [85, 92, 78, 88, 90, 75, 95, 82]
})
df
```

## Basic groupby

```python
# Average salary by city
df.groupby('City')['Salary'].mean()
```

## GroupBy wuth sum()

```python
# Total salary by city
df.groupby('City')['Salary'].sum()
```

## GroupBy with count()

```python
# Count employees by city
df.groupby('City')['Name'].count()
# Count not NaN values
```

## GroupBy with size()

```python
# Count rows in each group
df.groupby('City').size()
# Count all values even NaN
```

## Other Aggregation Functions

```python
# Minimum salary by city
df.groupby('City')['Salary'].min()
```

```python
# Maximum salary by city
df.groupby('City')['Salary'].max()
```

```python
# Median salary by city
df.groupby('City')['Salary'].median()
```

```python
# Standard diviation by city
df.groupby('City')['Salary'].std()
```

## Multiple aggregations

```python
df.groupby('City')['Salary'].agg(['min', 'max', 'mean'])
```

## .agg() with Named Results

```python
df.groupby('City')['Salary'].agg(Average_Salary = 'mean', Maximum_Salary = 'max', Minimum_Salary = 'min')
```

## Group Multiple Columns

```python
# Average age ang salary by city
df.groupby('City')[['Age', 'Salary']].mean()
```

## Multiple Aggregations on Multiple Columns

```python
df.groupby('City')[['Age', 'Salary']].agg(['min', 'max', 'mean', 'median', 'std'])
```

## GroupBy Two Columns

```python
df.groupby(['City', 'Department'])['Salary'].mean()
```

## Count Groups

```python
# Count employees by city and department
df.groupby(['City', 'Department']).size()
```

## Reset Index

```python
result = df.groupby('City')['Salary'].mean()
result
```

```python
# Cnvert index back to column
result = result.reset_index()
result
```

## GroupBy + Sort

```python
# Average salary by city
result = df.groupby('City')['Salary'].mean().sort_values(ascending = False)
result
```

## GroupBy + Filter

```python
# Calculate average salary
avg_salary = df.groupby('City')['Salary'].mean()
avg_salary[avg_salary > 3500]
```

## groupby().filter()

```python
# Keep groups with at least 3 employees
df.groupby('City').filter(lambda x: len(x) >= 3 )
```

## GroupBy with Diffrenr Functions

```python
# Multiple metrics by department
df.groupby('Department').agg({
    'Age': 'mean',
    'Salary': 'mean',
    'Score': 'max'
})
```

## Named Aggregation

```python
# Craet custom output names
result = df.groupby('Department').agg(
    Average_Name = ('Age', 'mean'),
    Average_Salary = ('Salary', 'mean'),
    Highest_Score = ('Score', 'max')
)
result
```

## GroupBy with nunique()

```python
# Count unique cities by department
df.groupby('Department')['City'].nunique()
```

## GroupBy with unique()

```python
# Count unique cities
df.groupby('Department')['City'].unique()
```

## GroupBy +  value_counts()

```python
# Count city occurences by department
df.groupby('Department')['City'].value_counts()
```

## A Real Data Analysis Example

```python
# Creat department summary
summary = df.groupby('Department').agg(Employees = ('Name','count'),
                                      Average_Age = ('Age', 'mean'),
                                      Average_Salary = ('Salary', 'mean'),
                                      Maximum_Saslary = ('Salary', 'max'),
                                      Average_Score = ('Score','mean')
                                      )
summary
```

## as_index = False

```python
# Keep group column as a normal column
df.groupby('City', as_index = False)['Salary'].mean()
```

```python
# Creat plotting-ready data
city_salary = df.groupby('City', as_index = False)['Salary'].mean()
city_salary
```
