## Creat DataFrame

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Reza", "Mina", "Omid"],
    "Age": [25, np.nan, 30, 28, np.nan],
    "Salary": [3000, 3500, np.nan, 2800, 3200],
    "City": ["Tehran", "Mashhad", np.nan, "Tabriz", "Tehran"]
})

df
```

## Detect Missing Values

```python
df.isna()
```

```python
df.isnull()
```

## Count Missing Values

```python
df.isna().sum()
```

## Check if DataFrame Has Missing Values

```python
df.isna().any()
```

```python
df.isna().any().any()
```

## Remove Rows with Misiing Values

```python
df_clean = df.dropna()

df_clean
```

## Remove Rows Based on Specific Column

```python
df_clean = df.dropna(subset = ['Salary'])

df_clean
```

## Fill Missing Values

```python
df['Age_new'] = df['Age'].fillna(0)
df
```

## Fill with Mean

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df
```

## Fill with Median

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df
```

## Fill Categorical Data

```python
df['city_un'] = df['City'].fillna('Unknown')
df
```

## Forward Fill

```python
df['City'] = df['City'].ffill()
df
```

## Backward Fill

```python
df['City'] = df['City'].bfill()
df
```
