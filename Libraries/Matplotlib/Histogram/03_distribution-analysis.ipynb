# Distribution Analysis & Customization

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Understanding Distribution

```python
# Generate normally distributed data
rng = np.random.default_rng(42)

data = rng.normal(
    loc = 50, 
    scale = 10,
    size = 1000
)

# Plot the histogram
plt.hist(data, bins = 20, edgecolor = 'black' )

plt.title('Data')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
```

## Identifying the Center

```python
mean = np.mean(data)
median = np.median(data)
print('mean: ', mean)
print('median: ', median)
```

## Identifying Spread

```python
std = np.std(data)
print('Standard diviasion: ', std)
```

## Comparing Two Distributions

```python
group_a = rng.normal(50, 8, 1000)

group_b = rng.normal(60, 8, 1000)

# Plot
plt.hist(
    group_a,
    bins = 20,
    alpha = 0.5,
    label = 'Group A'
)

plt.hist(
    group_b,
    bins = 20,
    alpha = 0.6,
    label = 'Group B'
)

plt.title('Comparision')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
```

## Using 'destiny = True' for Comparison

```python
plt.hist(
    group_a,
    bins = 20,
    alpha = 0.5,
    label = 'Group A'
)

plt.hist(
    group_b,
    bins = 20,
    alpha = 0.6,
    label = 'Group B'
)

plt.title('Comparision')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.legend()
plt.show()
```
