# Practical Data Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creat the Dataset

```python
rng = np.random.default_rng(42)

# Generate employee salaries
salaries = rng.normal(
    loc=5000,
    scale=1200,
    size=300
)

# Add some unusual values
salaries = np.append(salaries, [12000, 13000, 15000, 16000])

# Add missing values
salaries = np.append(salaries, [np.nan, np.nan])
```

## Data Cleaning

```python
# Remove missing values

df = salaries[~np.isnan(salaries)]

print('Original size: ', len(salaries))
print('Cleaned size: ', len(df))
```

# Basic Statistical Analysis

```python
mean = np.mean(df)
mean
```

```python
median = np.median(df)
median
```

```python
std = np.std(df)
std
```

## Crreat Histogram

```python
plt.hist(
    df,
    bins = 20, 
    edgecolor = 'black'
)

plt.title('Data')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
```

## Add Mean and Median

```python
plt.hist(
    df,
    bins = 20, 
    edgecolor = 'black'
)

plt.axvline(
    mean,
    linestyle = '--',
    label = 'Mean'
)

plt.axvline(
    median,
    linestyle = '-.',
    label = 'Median'
)

plt.title('Data')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.legend()

plt.show()
```

```python
# Too many bins
plt.hist(
    df,
    bins = 50,
    edgecolor = 'black'
)

plt.title('Data - 50 Bins')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
```

# Final Code

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate reproducible data
rng = np.random.default_rng(42)

# Generate employee salaries
salaries = rng.normal(
    loc=5000,
    scale=1200,
    size=300
)

# Add unusual values
salaries = np.append(
    salaries,
    [12000, 13000, 15000, 16000]
)

# Add missing values
salaries = np.append(
    salaries,
    [np.nan, np.nan]
)

# Remove missing values
salaries_clean = salaries[~np.isnan(salaries)]

# Calculate statistics
mean_salary = np.mean(salaries_clean)
median_salary = np.median(salaries_clean)
std_salary = np.std(salaries_clean)

print("Mean:", mean_salary)
print("Median:", median_salary)
print("Standard Deviation:", std_salary)
print("Minimum:", np.min(salaries_clean))
print("Maximum:", np.max(salaries_clean))

# Create the histogram
plt.hist(
    salaries_clean,
    bins=20,
    edgecolor="black",
    alpha=0.7
)

# Show mean and median
plt.axvline(
    mean_salary,
    linestyle="--",
    label="Mean"
)

plt.axvline(
    median_salary,
    linestyle="-.",
    label="Median"
)

# Add title and labels
plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

# Show legend
plt.legend()

plt.show()
```
