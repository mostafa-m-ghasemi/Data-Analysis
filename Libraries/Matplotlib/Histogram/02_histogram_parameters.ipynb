# Histogram Parameters

## Setting bins

```python
import numpy as np
import matplotlib.pyplot as plt

scores = np.array([
    45, 52, 67, 71, 73, 75, 78, 80, 81, 82, 84, 85, 85, 87, 88, 89, 90, 91, 92, 94
])
# Set the number of bins
plt.hist(scores, bins = 5)

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

## Using Custom Bin Edges

```python
bins = [40, 50, 60, 70, 80, 90, 100]

plt.hist(scores, bins = bins)

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

## Setting range

```python
plt.hist(scores, bins = 5,
        range = (40, 80))

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

## Using edgecolor

```python
plt.hist(scores, bins = 8,
        range = (40, 80),
        edgecolor = 'black')

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

## Using alpha

```python
plt.hist(scores, bins = 8,
        range = (40, 80),
        edgecolor = 'black',
        alpha = 0.7)

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

## Using density

```python
plt.hist(scores, bins = 8,
        range = (40, 80),
        edgecolor = 'black',
        alpha = 0.7,
        density = True)

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

```python
# density = False   --> Frequency
# density = True    --> Probability Destiny
```

## Using histtype

```python
plt.hist(scores, bins = 8,
        range = (40, 80),
        edgecolor = 'black',
        density = True,
        histtype = 'step')

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```

```python
# histtypes --> 'bar' , 'barstacked', 'step', 'stepfilled'
```
