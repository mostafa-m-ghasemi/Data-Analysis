# Histogram - Part 01

## Basic Histogram

```python
import numpy as np
import matplotlib.pyplot as plt

scores = np.array([
    45, 52, 67, 71, 73, 75, 78, 80, 81, 82, 84, 85, 85, 87, 88, 89, 90, 91, 92, 94
])

plt.hist(scores)
plt.show()
```

## Setting the Number of Bins

```python

plt.hist(scores, bins = 7)

plt.show()
```

## Compareing Diffrent Bin Sizes

```python
fig, ax = plt.subplots(2, 2, figsize = (8, 5))
ax[0, 0].set_title('Bin 5')
ax[0, 0].hist(scores, bins = 5)

ax[0, 1].set_title('Bin 10')
ax[0, 1].hist(scores, bins = 10)

ax[1, 0].set_title('Bin 15')
ax[1, 0].hist(scores, bins = 15)

ax[1, 1].set_title('Bin 20')
ax[1, 1].hist(scores, bins = 20)
plt.show
```

## Adding Labels and Title

```python
plt.hist(scores, bins = 18)

plt.title('Students_Scores')
plt.xlabel('Scores')
plt.ylabel('Frequency')

plt.show()
```
