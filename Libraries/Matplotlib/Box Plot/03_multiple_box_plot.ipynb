# Multiple Box Plot

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Multiple Data

```python
# Creat sample data
data1 = [12, 15, 14, 18, 20, 22, 19, 17, 16]
data2 = [20, 22, 25, 24, 28, 30, 26, 27, 29]
data3 = [8, 10, 12, 9, 14, 13, 11, 15, 10]
data = [data1, data2, data3]
```

## Multiole Box Plots

```python
# Creat multiple box plots
plt.boxplot(data)

plt.show()
```

## Add Labels

```python
plt.boxplot(data)

plt.xticks([1, 2, 3], ['Group A', 'Group B', 'Group C'])

plt.ylabel('Values')
plt.title('Comparison of Groups')

plt.show()
```

## Using labels

```python
plt.boxplot(data, tick_labels = ['Group A', 'Group B', 'Group C'])

plt.title('Comparison of Groups')

plt.show()
```

## Horizontal Multiple Box Plot

```python
plt.boxplot(data, vert = False, tick_labels = ['Group A', 'Group B', 'Group C'])

plt.title('Comparison of Groups')

plt.show()
```

## Box Plot with subplots

```python
fig, ax = plt.subplots(1, 2, figsize = (10, 4))

# First box plot
ax[0].boxplot(data)
ax[0].set_title('Vertical')

# Second box plot
ax[1].boxplot(data, vert = False)
ax[0].set_title('Horizontal')
plt.show()
```
