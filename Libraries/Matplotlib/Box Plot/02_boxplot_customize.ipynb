# Box Plot Customize

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Change box width

```python
data = [12, 15, 14, 18, 20, 22, 19, 17, 16, 25, 30, 21, 18, 15, 14]
```

```python
# Chang box width
plt.boxplot(data, widths = 0.5)

plt.show()
```

## Change Whisker Style

```python
# Customize whiskers
plt.boxplot(data, whiskerprops = dict(linestyle = '--'))

plt.show
```

## Customize Median

```python
# Customize median line
plt.boxplot(data, medianprops = dict(linewidth = 2))

plt.show()
```

## Customize Outliers

```python
# Customize outlier markers
plt.boxplot(data, flierprops = dict(marker = 's', markersize = 6))

plt.show()
```

## Customize the Box

```python
# Enable box customization
plt.boxplot(data, patch_artist = True)

plt.show()
```

```python
# Customize box appearance
plt.boxplot(data, patch_artist = True, boxprops = dict(facecolor = 'lightblue'))

plt.show()
```

## Complete Customization Example

```python
data = [12, 15, 14, 18, 20, 22, 19, 17, 16, 25, 30, 21, 18, 15, 14]

plt.figure(figsize = (8, 5))

plt.boxplot(data,
            widths = 0.5,
            patch_artist = True,
            boxprops = dict(facecolor = 'lightblue'),
           medianprops = dict(linewidth = 2.5),
           flierprops = dict(marker = 's', markersize = 8))

plt.title('Customize Box Plot')
plt.ylabel('Values')


plt.show()
```
