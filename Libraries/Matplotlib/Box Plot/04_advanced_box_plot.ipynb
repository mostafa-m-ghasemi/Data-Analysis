# Box Plot Advanced

```python
import numpy as np
import matplotlib.pyplot as plt

data = [12, 15, 14, 18, 20, 22, 19, 17, 16, 25, 30, 21, 18, 15, 14]
```

## Show Mean

```python
plt.boxplot(data, showmeans = True)

plt.show()
```

## Customize Mean

```python
plt.boxplot(data, showmeans = True, meanprops = dict(marker = 'D', markersize = 8))

plt.show()
```

## Show Mean as a Line

```python
plt.boxplot(data, showmeans = True, meanline = True)

plt.show()
```

## Notched Box Plot

```python
# confidence interval around the median
plt.boxplot(data, notch = True)

plt.show()
```

## Hide Outliers

```python
plt.boxplot(data, showfliers = False)

plt.show()
```

## Whisker Range

```python
plt.boxplot(data, whis = 0.8)

plt.show()
```

## Finak Example

```python
data1 = [12, 15, 14, 18, 20, 22, 19, 17, 16, 25, 30]
data2 = [20, 22, 25, 24, 28, 30, 26, 27, 29]
data3 = [8, 10, 12, 9, 14, 13, 11, 15, 10]

data = [data1, data2, data3]
```

```python
fig, ax = plt.subplots(figsize = (9, 5))

ax.boxplot(data,
          tick_labels = ['Group A', 'Group B', 'Group C'],
          patch_artist = True,
          showmeans = True,
          notch = True,
          boxprops = dict(facecolor = 'lightblue'),
          medianprops = dict(linewidth = 2.5),
          meanprops = dict(marker = 'D', markersize = 5),
          whiskerprops = dict(linestyle = '--'),
          flierprops = dict(marker = 'o', markersize = 5))
ax.set_title('Box Plot Advenced')
ax.set_ylabel('Values')

plt.show()
```
