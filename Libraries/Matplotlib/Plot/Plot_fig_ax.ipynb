```python
import matplotlib.pyplot as plt
import numpy as np
```

# FIG + AX

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 5, 3, 8, 6, 9]
```

## 2 plots

```python
fig, ax = plt.subplots(
    2, 1,
    figsize = (8,8)
)

ax[0].plot(x,y)
ax[1].plot(y,x)

plt.show()
```

```python
# Options
```

```python
fig, ax = plt.subplots(
    2,1,
    figsize = (8,8)
)
ax[0].plot(x,y,
        color = 'b',
        marker = 'o')
ax[1].plot(x,y,
           color = 'r',
           linestyle = '--',
           marker = 's')

ax[0].set_title('Plot 1')
ax[1].set_title('Plot 2')

ax[0].set_xlabel('X')
ax[0].set_ylabel('Y')

ax[1].set_xlabel('X')
ax[1].set_ylabel('Y')

ax[0].grid()
ax[1].grid(alpha = 0.5)

plt.tight_layout()
plt.show()
```

## 4 Plots

```python
# 4 plots

fig, ax = plt.subplots(
    2, 2,
    figsize=(10, 8)
)

ax[0, 0].plot(x, y)
ax[0, 1].plot(x, y)
ax[1, 0].plot(x, y)
ax[1, 1].plot(x, y)

plt.show()
```

```python
# 4 plots - complete

fig, ax = plt.subplots(
    2, 2,
    figsize=(10, 8)
)

ax[0, 0].plot(
    x, y,
    color="blue",
    marker="o"
)

ax[0, 1].plot(
    x, y,
    color="red",
    linestyle="--"
)

ax[1, 0].plot(
    x, y,
    color="green",
    marker="s"
)

ax[1, 1].plot(
    x, y,
    color="black",
    linestyle=":"
)

ax[0, 0].set_title("Plot 1")
ax[0, 1].set_title("Plot 2")
ax[1, 0].set_title("Plot 3")
ax[1, 1].set_title("Plot 4")

for a in ax.flat:
    a.grid(alpha = 0.3)

plt.tight_layout()

plt.show()
```
