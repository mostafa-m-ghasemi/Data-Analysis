```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data
x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
y = [2, 5, 3, 8, 6, 9, 7, 4, 9]
```

# Basic plot

```python
plt.plot(x, y)
plt.show()
```

# Line

```python
plt.plot(x, y,
        color = 'green',
        linestyle = '-.',
        linewidth = 3)
# common linestyles = '-' , '--' , ':' , '-.'
plt.show()
```

# Marker

```python
plt.plot(x, y,
        color = 'green',
         # you can use 'r', 'b', ....
        linestyle = '-.',
        linewidth = 2,
        marker = 'o')
# common Markes = 'o' , 's' , '^' , 'D' , '*' , '+' , 'x' , '.'
plt.show()
```

# Marker options

```python
plt.plot(x, y,
        marker = 'o', markersize = 10, markerfacecolor = 'yellow', markeredgecolor = 'black', markeredgewidth = 3)

plt.show()
```

# Format string

```python
# Format string: combines color, marker, and line style in one short string
# Order: [color][marker][linestyle]
# Example : 'ro--' = red + o + --
```

```python
plt.plot(x, y, 'yo-.')

plt.show()
```

# Alpha

```python
# Alpha : controls the transparency of the plot
# Range : 0 to 1
# 0 = fully transparent , 1 = fully opaque
# Example : alpha = 0.5 means 50% transparency
```

```python
plt.plot(x, y, alpha = 0.5)

plt.show()
```

# Title and labels

```python
plt.plot(x, y)
plt.title('Main Title')
plt.xlabel('---X---')
plt.ylabel('---Y---')

plt.show()
```

# Grid

```python
plt.plot(x, y)
plt.grid(
    linestyle = '--',
    linewidth = 0.7,
    alpha = 0.3
)

plt.show()
```

# Limits

```python
plt.plot(x, y)
plt.xlim(0, 10)
plt.ylim(0, 10)

plt.show()
```

# Ticks

```python
# Ticks : Controls the positions and labels shown on the axes.
# You can set custom positions, labels, rotation and font size
# Example : plt.xticks([1, 2, 3],['A','B','C'], rotation = 45)
```

```python
plt.plot(x, y)
plt.xticks([1, 2, 3, 4, 5, 6, 7, 8, 9], rotation = 45)
plt.yticks([0, 2, 4, 6, 8, 10, 12, 14, 16], rotation = 45)

plt.show()
```

# Legend

```python
# Legend: Identifies each plotted line using its label.
# Add label inside plot(), then call legend() to display it.
# loc control the legend position
```

```python
z =[0, 2, 4, 6, 8, 10, 12, 14, 16]

plt.plot(x, y, label = 'First')
plt.plot(z, y, label = 'Second')

plt.legend(
    loc = 'upper left',
    fontsize = 10,
    frameon = True
)
plt.show()
```

# Text

```python
# Text : adds custom text at a specific position on the plot.
# You can also change fontsize, color, rotation, and alignment
```

```python

plt.plot(x, y)
plt.text(4, 8, 'Important')

plt.show()
```

# Annotation

```python
# Annotayion : adds a text label with an arrow pointing to a specific point
# You can customiza the arrow, fontsiza, color, and alignment
```

```python
plt.plot(x, y)

plt.annotate(
    "Highst",
    xy = (9, 9),
    xytext = (4, 7),
    arrowprops = dict(arrowstyle = '->')
)
```

# Complete Plot

```python
plt.figure(figsize = (10, 6))

plt.plot(
    x, y,
    color = 'darkblue',
    linestyle = '-',
    linewidth = 2,
    marker = 'o',
    markersize = 8,
    markerfacecolor = 'y',
    markeredgewidth = 1.5,
    markeredgecolor = 'black',
    alpha = 0.9,
    label = 'Label'
)

plt.title('Report',
    fontsize = 16)
plt.xlabel('Xlabel',
          fontsize = 12)
plt.ylabel('Ylabel',
          fontsize = 12)

plt.xlim(0, 10)
plt.ylim(0, 10)

plt.xticks([1, 2, 3, 4, 5, 6, 7, 8, 9])
plt.yticks([0, 2, 4, 6, 8, 10, 12, 14, 16])

plt.grid(
    linestyle = '--',
    alpha = 0.4
)

plt.legend(loc = 'upper left')

plt.tight_layout()

plt.show()
```
