# Basic Bar Chart

```python
import matplotlib.pyplot as plt
```

```python
# Data
x = ["A", "B", "C", "D"]
y = [10, 25, 15, 30]
```

```python
# Create figure and axes
fig, ax = plt.subplots(figsize=(8, 5))

plt.show()
```

```python
# Create bar chart
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(
    x,
    y,
    width=0.7,
    align="edge",
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    alpha=0.8
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    x,
    y,
    width=0.7,
    align="center",
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    alpha=0.8
)

# Title
ax.set_title("Sales by Product")

# Axis labels
ax.set_xlabel("Product")
ax.set_ylabel("Sales")

# X ticks
ax.set_xticks(range(len(x)))
ax.set_xticklabels(x)

# Rotate x-axis labels
ax.tick_params(axis="x", rotation=45)

# Axis limits
ax.set_xlim(-0.5, 3.5)
ax.set_ylim(0, 35)

# Show plot
plt.show()
```
