```python
import matplotlib.pyplot as plt
```

# Grid

```python
x = ["A", "B", "C", "D"]
y = [10, 25, 15, 30]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x, y)

# Add horizontal grid
ax.grid(axis="y", alpha=0.3)

# Put grid behind bars
ax.set_axisbelow(True)

plt.show()
```

# Bar Labels

```python

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x, y)

# Show values on bars
ax.bar_label(bars)

plt.show()
```

# Bar Label Padding

```python
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x, y)

# Add space between bar and label
ax.bar_label(bars, padding=4)

plt.show()
```


# Format Bar Labels


```python
x = ["A", "B", "C", "D"]
y = [10.25, 25.50, 15.75, 30.80]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x, y)

# Format values
ax.bar_label(
    bars,
    fmt="%.1f",
    padding=3
)

plt.show()
```

#  Legend

```python
fig, ax = plt.subplots(figsize=(8, 5))

# Add label for legend
bars = ax.bar(
    x,
    y,
    label="Sales"
)

# Show legend
ax.legend()

plt.show()
```


#  Horizontal Bar Chart

```python

fig, ax = plt.subplots(figsize=(8, 5))

# Create horizontal bars
bars = ax.barh(x, y)

# Show values
ax.bar_label(bars, padding=3)

plt.show()
```


#  Sort Bars - Ascending


```python
data = {
    "A": 30,
    "B": 10,
    "C": 25,
    "D": 15
}

# Sort data
data_sorted = dict(
    sorted(data.items(), key=lambda item: item[1])
)

x_sorted = list(data_sorted.keys())
y_sorted = list(data_sorted.values())

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x_sorted, y_sorted)

ax.bar_label(bars, padding=3)

plt.show()
```

```python
#  Sort Bars - Descending

data_sorted = dict(
    sorted(
        data.items(),
        key=lambda item: item[1],
        reverse=True ## --> important
    )
)

x_sorted = list(data_sorted.keys())
y_sorted = list(data_sorted.values())

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(x_sorted, y_sorted)

ax.bar_label(bars, padding=3)

plt.show()
```

#  Text

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(x, y)

# Add text at specific position
ax.text(
    2,
    27,
    "Highest",
    ha="center"
)

plt.show()
```

#  Annotate

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(x, y)

# Add annotation with arrow
ax.annotate(
    "Highest",
    xy=(3, 30),
    xytext=(2, 35),
    arrowprops=dict(arrowstyle="<-")
)

plt.show()
```

#  Spines

```python

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(x, y)

# Remove top and right spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.show()
```

#  Figure Size

```python

# Set figure size
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(x, y)

plt.show()
```

# DPI

```python
# Set resolution
fig, ax = plt.subplots(
    figsize=(10, 6),
    dpi=60

)

ax.bar(x, y)

plt.show()
```

#  Save Figure

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(x, y)

# Save chart
fig.savefig(
    "bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
