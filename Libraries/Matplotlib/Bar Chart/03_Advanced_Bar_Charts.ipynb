# Advanced Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Grouped Bar Chart

```python
products = ['A','B','C','D']
sales_2025 = [20, 30, 25, 40]
sales_2026 = [25, 35, 30, 45]
```

```python
# Creat x positions
x = np.arange(len(products))

# Set bar width
width = 0.35

fig, ax = plt.subplots(figsize = (10, 6))

# Creat first group
bars1 = ax.bar(
    x - width / 2,
    sales_2025,
    width,
    label = '2025'
)

# Crerat second group

bars2 = ax.bar(
    x + width /2,
    sales_2026,
    width,
    label = '2026'
)

# Set x-axis labels
ax.set_xticks(x)
ax.set_xticklabels(products)

# Add title and labels
ax.set_title('Sales Compersion')
ax.set_xlabel('Products')
ax.set_ylabel('Sales')

# Add legend
ax.legend()


plt.show()
```

## Grouped Bar + Labels

```python
fig, ax = plt.subplots(figsize = (10, 6))
bars1 = ax.bar(
    x - width / 2,
    sales_2025,
    width,
    label = '2025'
)

bars2 = ax.bar(
    x + width /2,
    sales_2026,
    width,
    label = '2026'
)

# Show values
ax.bar_label(bars1, padding = 3)
ax.bar_label(bars2, padding = 3)

ax.set_xticks(x)
ax.set_xticklabels(products)

ax.set_title('Sales Compersion')
ax.set_xlabel('Products')
ax.set_ylabel('Sales')

ax.legend()


plt.show()
```

## Stacked Bar Chart + Labels

```python
online = [20, 30, 25, 40]
store = [10, 15, 20, 10]

fig, ax = plt.subplots(figsize = (10, 6))

# First Layer
bars1 = ax.bar(
    products,
    online,
    label = 'Online'
)

# Second Layer
bars2 = ax.bar(
    products,
    store,
    label = 'store'
)

# Show values
ax.bar_label(bars1, padding = 3)
ax.bar_label(bars2, padding = 3)
# Add legend
ax.legend()

plt.show()
```

## Mulpile Stacked Groups

```python
other = [5, 8, 10, 7]

fig, ax = plt.subplots(figsize = (10, 6))

# First Layer
bars1 = ax.bar(
    products,
    online,
    label = 'Online'
)

# Second Layer
bars2 = ax.bar(
    products,
    store,
    label = 'store'
)

# Calculate bottom for third layer
bottom = np.array(online) + np.array(store)

# Third layer
ax.bar(
    products,
    other,
    bottom = bottom,
    label = 'other'
)

ax.set_title('Total Sales')
ax.set_xlabel('Products')
ax.set_ylabel('Sales')

ax.legend()
plt.show()
```

## Peercentage Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
percentage = [25, 40, 20, 15]

fig, ax = plt.subplots(figsize = (8, 5))

bars = ax.bar(categories, percentage)

# Show percentage values
ax.bar_label(
    bars,
    fmt = '%.0f%%'
)

ax.set_title('Percentage')
ax.set_xlabel('Ctegory')
ax.set_ylabel('Percentage (%)')

ax.set_ylim(0, 50)

plt.show()
```

## Horizontal Percentage Bar Chart

```python
fig, ax = plt.subplots(figsize = (8, 5))

bars = ax.barh(
    categories,
    percentage
)

# Show values
ax.bar_label(
    bars,
    fmt = '%.0f%%',
    padding = 3
)

ax.set_title('Percentage')
ax.set_xlabel('Ctegory')
ax.set_ylabel('Percentage (%)')


plt.show()
```

# Complete Bar Chart Template

```python
categories = ['A', 'B', 'C', 'D', 'E', 'F']
values = [10, 25, 15, 30, 10, 35]

fig, ax = plt.subplots(figsize = (10, 6))

# Creat bars
bars = ax.bar(
    categories,
    values,
    color = 'skyblue',
    edgecolor = 'black',
    linewidth = 1,
    alpha = 0.8
)

# Title
ax.set_title('Value by Categories')

# Axis label
ax.set_xlabel('Categories')
ax.set_ylabel('Values')

# Grid
ax.grid(
    axis = 'y',
    alpha = 0.3
)

# Put grid behind bars
ax.set_axisbelow(True)

# Show values
ax.bar_label(
    bars,
    color = 'darkblue'
)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Save figure
fig.savefig('bar_chart.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```
