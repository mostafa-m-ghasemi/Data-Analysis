# Scatter Plot

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7, 8]
score = [50, 55, 60, 65, 72, 78, 85, 90]
```

## Change Point Color

```python
fig, ax = plt.subplots()

ax.scatter(hours, score, color = 'red')

plt.show()
```

```python
# Use c instead of color
fig, ax = plt.subplots()

ax.scatter(hours, score, c = 'brown')

plt.show()
```

## Add Edge Color

```python
fig, ax = plt.subplots()

ax.scatter(hours, score,
           c = 'brown',
          edgecolor = 'skyblue')

plt.show()
```

## Combime propertises

```python
fig, ax = plt.subplots()

ax.scatter(hours, score,
           marker = 'D',
           s = 90,
           c = 'brown',
          edgecolor = 'skyblue',
          alpha = .9)

plt.show()
```
