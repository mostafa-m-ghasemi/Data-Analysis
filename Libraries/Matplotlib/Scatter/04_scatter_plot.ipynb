# Scatter Plot

## Creat a Color Variable

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5, 6, 7, 8]
score = [50, 55, 60, 65, 72, 78, 85, 90]
age = [18, 19, 20, 21, 22, 23, 24, 25]
```

## Color Points Based on Valuea

```python
fig, ax = plt.subplots()
scatter = ax.scatter(
    hours,
    score,
    c = age,
    cmap='viridis'
)

plt.show()
```

## Add Colorbar

```python
fig, ax = plt.subplots()
scatter = ax.scatter(
    hours,
    score,
    c = age,
    cmap='viridis'
)
fig.colorbar(scatter, ax = ax, label = 'Age')

plt.show()
```

## Try Diffrent Colormaps

```python
fig, ax = plt.subplots()
scatter = ax.scatter(
    hours,
    score,
    c = age,
    cmap='magma'
)
fig.colorbar(scatter, ax = ax, label = 'Age')

plt.show()
```
