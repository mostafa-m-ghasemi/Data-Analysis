## Figure Size, Grid & Appearance

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4 ,5 ,6 ,7, 8]
score = [50, 55, 60, 65, 72, 78, 85, 90]

fig, ax = plt.subplots(figsize = (8, 5))

# creat scatter plot
ax.scatter(
    hours,
    score,
    s = 100,
    alpha = 0.7,
    edgecolor = 'black'
)

# Add labels and titles
ax.set_xlabel('Hours')
ax.set_ylabel('Scores')
ax.set_title('Title')

# Ad Grid
ax.grid(True, alpha = 0.3)
plt.show
```

## Annotations $ Highlighting Points

```python
fig, ax = plt.subplots(figsize = (8, 5))

ax.scatter(
    hours,
    score,
    s = 100,
    alpha = 0.7
)

# Highlight and important points

ax.scatter(
    8,
    90,
    s = 150,
    color = 'r'
)

# Add a text to the point
ax.annotate(
    'Highest Score',
    xy = (8, 90),
    xytext = (5.5, 88),
    arrowprops = dict(arrowstyle = '->')
)

ax.grid(True, alpha = 0.2)

plt.show()
```

## Complete Example

```python
age = [18, 19, 20, 21, 22, 23, 24, 25]



fig, ax = plt.subplots(figsize = (8, 5))

# Use age to control points color

scatter = ax.scatter(
    hours,
    score,
    c = age,
    s = 100,
    alpha = 0.8,
    edgecolor = 'black'
)

# Add color bar
fig.colorbar(scatter, ax = ax, label = 'Age')

# Add title and labels
ax.set_title('Title')
ax.set_xlabel('Hours')
ax.set_ylabel('Exam Scores')

# Add grid
ax.grid(True, alpha = 0.3)

# Highlight
ax.annotate(
    'Highest Score',
    xy = (8, 90),
    xytext = (5, 90),
    arrowprops = dict(arrowstyle = '->')
)

plt.show()
```
