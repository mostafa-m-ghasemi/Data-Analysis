# Scatter Plot

## Creat Two Groups

```python
import matplotlib.pyplot as plt

hours_a = [1, 2, 3, 4, 5]
score_a = [50, 55, 60, 65, 72]

hours_b = [4, 5, 6, 7, 8]
score_b = [65, 72, 78, 85, 90]
```

## Plot Multiple Groups

```python
fig, ax = plt.subplots()

# Plot the first group
ax.scatter(hours_a, score_a, label = 'Group A')

# Plot second group
ax.scatter(hours_b, score_b, label = 'Group B')

ax.legend()

plt.show()
```

## Customize Each Griup

```python
fig, ax = plt.subplots()

# Plot the first group
ax.scatter(hours_a,
           score_a,
           label = 'Group A',
          marker = 'o',
          s = 100,
          color = 'b')

# Plot second group
ax.scatter(hours_b,
           score_b,
           label = 'Group B',
          marker = '^',
          s = 100,
          color = 'r')

ax.legend()

plt.show()
```
