# Scatter Plot

```python
import matplotlib.pyplot as plt
```

## Creat Data

```python
# Creat x and y data
hours = [1, 2, 3, 4, 5, 6, 7, 8]
score = [50, 55, 60, 65, 72, 78, 85, 90]
```

## Basic Scatter Plot

```python
# Plot points without connection them
plt.scatter(hours, score)

plt.show()
```

## Add Labels and Title

```python
# Creat scatter plot
plt.scatter(hours, score)

# Add axis labels
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')

# Add title
plt.title('Study Hours vs Exam Score')

plt.show()
```

## Using Figure and Axe

```python
fig, ax = plt.subplots()


# Plot scatter points
ax.scatter(hours, score)

# Add labels and title
ax.set_title('Study Hours vs Exam Score')
ax.set_xlabel('Hours Studies')
ax.set_ylabel('Exam Score')


plt.show()
```
