# Scatter Plot

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7, 8]
score = [50, 55, 60, 65, 72, 78, 85, 90]
```

## Change Marker and Marker Size

```python
fig, ax = plt.subplots()

# Change the shape of points
ax.scatter(hours, score, marker = '^', s =100)

plt.show()
```

## Change Transparency

```python
fig, ax = plt.subplots()


ax.scatter(hours, score, alpha = 0.5)

plt.show()
```

## Example

```python
fig, ax = plt.subplots()


ax.scatter(hours, score,
           alpha = 0.5,
          marker = 'x',
          s = 70)

plt.show()
```
