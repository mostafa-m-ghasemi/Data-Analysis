# Basic Box Plot

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sample data
data = [12, 15, 14, 18, 20, 22, 19, 17, 16, 25, 30, 21, 18, 15, 14]

# Creat a basic box plot
plt.boxplot(data)

plt.show()
```

## Horizontal Box Plot

```python
plt.boxplot(data, vert = False)

plt.show()
```

## Change the Figure Size

```python
plt.figure(figsize = (8, 4))
plt.boxplot(data)

plt.show()
```

## Add a Title

```python
plt.figure(figsize = (8, 4))
plt.boxplot(data)

plt.title('Box Plot')

plt.show()
```

## Add Labels

```python
plt.figure(figsize = (8, 4))
plt.boxplot(data)

plt.title('Box Plot')
plt.ylabel('Values')

plt.show()
```
