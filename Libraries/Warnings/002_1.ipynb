# Why is it safer to silence only ConvergenceWarning instead of silencing everything with a bare ignore?

## 001 - Divide by zero or log of negative means NaN in data, model learns garbage.

```python
import warnings
import numpy as np

# show all warnings for demo
warnings.filterwarnings("always")

# risky math creates NaN and inf
values = np.log(np.array([10, 0, -5]))
print(values)
```

## 002 - Common real bug in sklearn. Fit with dataframe, predict with nameless array.

```python
import warnings
import pandas as pd
from sklearn.linear_model import LogisticRegression

# show all warnings for demo
warnings.filterwarnings("always")

# train with named columns
train_df = pd.DataFrame({"age": [20, 30, 40, 50], "score": [1, 2, 3, 4]})
model = LogisticRegression()
model.fit(train_df, [0, 0, 1, 1])

# predict without names triggers UserWarning
model.predict([[35, 2]])
```

## 003 - This says result will change in next version. Silencing it is technical debt

```python
import warnings

# warn about future breaking change
warnings.filterwarnings("always")

warnings.warn("default solver will change in next version", FutureWarning)
print("code still works today")
```

# Safe rule: silence only the warning you know by name, not all

```python
import warnings
from sklearn.exceptions import ConvergenceWarning

# safe: silence only one known noisy category
warnings.filterwarnings("ignore", category=ConvergenceWarning)
```
