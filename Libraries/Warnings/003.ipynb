# Temporary Control with catch_warnings() and Custom Categories

```python
import warnings

# normal behavior everywhere
warnings.filterwarnings("always")

# temporary ignore only inside block
with warnings.catch_warnings():
    warnings.simplefilter("ignore", category=FutureWarning)
    warnings.warn("hidden inside block", FutureWarning)
    print("inside: no FutureWarning shown")

# outside block warning is visible again
warnings.warn("visible outside block", FutureWarning)
```

```python
import warnings

# capture warnings into a list
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    warnings.warn("first issue", UserWarning)
    warnings.warn("second issue", FutureWarning)
    print("total caught:", len(caught))
    print("first message:", str(caught[0].message))
```

Sometimes you want to warn yourself, like train-test leak. Make a new class from UserWarning.

```python
import warnings

# define custom warning type
class DataLeakWarning(UserWarning):
    pass

def train_model(leak=False):
    # warn if risky case found
    if leak:
        warnings.warn("train data leaked into test", DataLeakWarning)
    return "model done"

# always show this custom type
warnings.filterwarnings("always", category=DataLeakWarning)

print(train_model(leak=True))
```

```python
import warnings

# ignore only custom warning, keep others
warnings.filterwarnings("ignore", category=DataLeakWarning)

print(train_model(leak=True))
print("custom warning hidden")
```

# Pro rule: global only for warning you know by name, temporary with catch_warnings for code not yours, custom for case not fatal yet.
