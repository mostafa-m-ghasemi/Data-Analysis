# Suppressing ML Warnings with filterwarnings

```python
import warnings

# show warning every time
warnings.filterwarnings("always")

warnings.warn("future change in sklearn", FutureWarning)
warnings.warn("future change in sklearn", FutureWarning)
```

```python
import warnings

# ignore all warnings
warnings.filterwarnings("ignore")

warnings.warn("you will not see this", FutureWarning)
print("no warning shown")
```

```python
import warnings

# ignore only FutureWarning
warnings.filterwarnings("ignore", category=FutureWarning)

warnings.warn("hidden warning", FutureWarning)
warnings.warn("still visible", UserWarning)
```

```python
import warnings
from sklearn.exceptions import ConvergenceWarning

# ignore only convergence warnings from sklearn
warnings.filterwarnings("ignore", category=ConvergenceWarning)

print("convergence warnings off")
```

```python
import warnings

# reset to default behavior
warnings.filterwarnings("default")

print("warnings back to normal")
```
