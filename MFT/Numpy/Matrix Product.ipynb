```python
import numpy as np
```

## Vector

```python
vec01 = np.random.randint(1, 10, 10)
vec02 = np.random.randint(1, 10, 10)
print(vec01)
print(vec02)
```

## Hadamard

```python
vec01 * vec02
```

## Inner

```python
np.inner(vec01, vec02)
```

## Dot

```python
np.dot(vec01, vec02)
```

## Matrix

```python
mat01 = np.array([[1, 2] , [3, 4]])
mat02 = np.array([[5, 6], [7, 8]])
print(mat01)
print(mat02)
```

## Hadamard --> array * array

```python
mat01 * mat02
```

## Inner --> Row * Row

```python
np.inner(mat01, mat02)
```

## Dot --> Row * Column

```python
np.dot(mat01, mat02)
```
