```python
import numpy as np
```

```python
vec1 = np.arange(1, 10)
print(vec1)
vec2 = np.arange(11, 20)
print(vec2)
```

## hstack Vectors

```python
np.hstack((vec1, vec2))
```

## Vstack Vectors

```python
np.vstack((vec1, vec2))
```

## c_ Vectors

```python
np.c_[vec1, vec2]
```

## hstack Matrix

```python
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.array([[7, 8, 9], [10, 11, 12]])
print(array1)
print('-' *10)
print(array2)
```

```python
np.hstack((array1, array2))
```

## vstack Matrix

```python
np.vstack((array1, array2))
```

## Concatenate

```python
np.concatenate((array1, array2), axis=1)
```

```python
np.concatenate((array1, array2), axis=0)
```
