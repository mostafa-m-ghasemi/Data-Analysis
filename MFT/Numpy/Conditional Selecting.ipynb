```python
import numpy as np
```

# Conditional Selecting 

```python
scores = np.random.randint(5,20,(10,4))
scores
```

## Masking

```python
scores[True] # --> Select everything
```

```python
scores[[False, True, True, True, True, True, True, True, True, False]] # Select rows when True
```

```python
mask = scores > 17 # return True if condition is true in matrix
mask
```

```python
scores[mask] # return only True values
```

```python
scores[scores > 17] # put the mask inside [] of matrix
```

```python
mask = (scores > 14) & (scores <= 17) # The () are really important, it doesn't work without them!
scores[mask]

 # or we can use:   scores[(scores > 14) & (scores <= 17)]
```
