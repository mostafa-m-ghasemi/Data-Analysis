## Random

```python
import numpy as np
```

### random

```python
array = np.random.random((4, 10)) (row, col) making random number betweeon 0 and 1 --> 1 is not included
array
```

### rand

```python
array = np.random.rand(10, 4) # (row, col) making random number betweeon 0 and 1 --> 1 is not included
array
```

```python
vector = np.random.rand(10) # making an one row array y random numbers between 0 and 1 --> 1 is not included
vector
```

### randint

```python
array = np.random.randint(0, 20) # make a random number
array
```

```python
array = np.random.randint(0, 20, 10) # make random number more than 1 --> (start, stop-1, how many number do you need!)
array
```

```python
array = np.random.randint(0, 20, (5,5)) # make a matrix of random numbers
array
```
