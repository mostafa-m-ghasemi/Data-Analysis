```python
import numpy as np
```

```python
scores = np.random.randint(5, 20, (10, 4))
scores
```

```python
conditions = [
    (scores >= 17) & (scores <= 20),
    (scores >= 14) & (scores < 17),
    (scores >= 10) & (scores < 14),
    (scores < 10)
]
selection = [
    "Grade A",
    "Grade B",
    "Grade C",
    "Grade F"
]
grade = np.select(conditions, selection, default="np.nan")
grade
```
